###**5. DATA HANDLING**

In the following section is showed how the datasets employed in the work are constructed which are both instance of Maximum-Weight Independent Set problem:
- 3-SAT (decision version)
- maximum clique (optimization version)

Formally, the Maximum-Weight Independent Set problem is defined as  follows:

Input: An undirected graph $G(=(\mathrm{V}(G), \mathrm{E}(G)))$, where each vertex $i \in \mathrm{V}(G)=\{1, \ldots, n\}$ is weighted by a positive rational number $c_i$

Output: A subset $S \subseteq \mathrm{V}(G)$ such that $S$ is independent (i.e., for each $i, j \in \mathrm{V}(G), i \neq j, i j \notin \mathrm{E}(G)$ ) and the total weight of $S\left(=\sum_{i \in S} c_i\right)$ is maximized. Denote the optimal set by mis $(G)$.
This is referred as the optimization version of MIS. The corresponding decision problem is:
Question: Is there an independent subset $S \subseteq \mathrm{V}(G)$ such that the total weight is at least $k$ ? (where the positive rational number $k$ is the extra input parameter.)

Technically speaking, the decision version of MIS is NP-complete, and the optimization version of MIS is NP-hard.

One way to solve the MIS problem is by maximizing a quadratic binary function $\mathcal{Y}$ (also known as pseudoboolean function) defined in the following theorem
Theorem 2.1. If $J_{i j} \geq \min \left\{c_i, c_j\right\}$ for all $i j \in \mathrm{E}(G)$, then the maximum value of
$$
\mathcal{Y}\left(x_1, \ldots, x_n\right)=\sum_{i \in \mathrm{V}(G)} c_i x_i-\sum_{i j \in \mathrm{E}(G)} J_{i j} x_i x_j
$$
is the total weight of the MIS. In particular if $J_{i j}>\min \left\{c_i, c_j\right\}$ for all $i j \in \mathrm{E}(G)$, then $\operatorname{mis}(G)=\{i \in \mathrm{V}(G)$ : $\left.x_i^*=1\right\}$, where $\left(x_1^*, \ldots, x_n^*\right)=\arg \max _{\left(x_1, \ldots, x_n\right) \in\{0,1\}^n} \mathcal{Y}\left(x_1, \ldots, x_n\right)$.

Notice that in this formulation, we only require $J_{i j}>\min \left\{c_i, c_j\right\}$, and thus there is freedom in choosing this parameter. It is easy to see (by changing the variables $x_i=\frac{1+s_i}{2}$ ) that MIS is equivalent to minimizing the Ising energy function :
$$
\mathcal{E}\left(s_1, \ldots, s_n\right)=\sum_{i \in \mathrm{V}(G)} h_i s_i+\sum_{i j \in \mathrm{E}(G)} J_{i j} s_i s_j,
$$
which is the eigenfunction of the following Ising Hamiltonian:
$$
\mathcal{H}_C=\sum_{i \in \mathrm{V}(G)} h_i \sigma_i^z+\sum_{i j \in \mathrm{E}(G)} J_{i j} \sigma_i^z \sigma_j^z
$$
where $h_i=\sum_{j \in \mathrm{nbr}(i)} J_{i j}-2 c_i, \operatorname{nbr}(i)=\{j: i j \in \mathrm{E}(G)\}$, for $i \in \mathrm{V}(G)$.

#####**5.1.2 Global functions**

necessary to fill missing edges with -1 values. This make sense given -1 value is not a valid value for a real edge features

```python
def sample_graph(min_embeddable: int = MIN_EMBEDDABLE, max_embeddable: int = MAX_EMBEDDABLE,
                 min_edge_prob: int = MIN_EDGE_PROB, max_edge_prob: float = MAX_EDGE_PROB,
                 random_seed: int = RANDOM_SEED, minimum_torque_value: float = MINIMUM_TORQUE_VALUE,
                 maximum_torque_value: int = MAXIMUM_TORQUE_VALUE, minimum_anneal_time: int = MINIMUM_ANNEAL_TIME,
                 maximum_anneal_time: int = MAXIMUM_ANNEAL_TIME, from_erdos: bool = False, sat: bool = False) -> List[Union[Dict[str, Union[float, int]], torch_geometric.data.Data]]:
    '''
    sample a graph returning input and embed graph features and data item
    '''
    if sat:
        # selecting random sat
        random_instance: int = random.choice(DIMACS_CONTAINER)
        DIMACS_CONTAINER.remove(random_instance)
        # construct related graph
        complement_graph: networkx.Graph = networkx.Graph()
        incremental: int = 0
        external_edge_name_list: List[str] = []
        for triplets in random_instance["list_cnf"]:
            index: int = 0
            edge_name_list: List[str] = []
            for elem in triplets:
                complement_graph.add_node(str(incremental) + "_" + str(elem))
                edge_name_list.append(str(incremental) + "_" + str(elem))
                external_edge_name_list.append(str(incremental) + "_" + str(elem))
                incremental += 1
            for i in range(len(edge_name_list)):
                complement_graph.add_edge(edge_name_list[i], edge_name_list[(i+1)%3])
        for elem in external_edge_name_list:
            split: List[str] = elem.split("_")[1]
            for elem1 in external_edge_name_list:
                split1: List[str] = elem1.split("_")[1]
                if -abs(int(split)) == int(split1):
                    complement_graph.add_edge(elem, elem1)
        m: float = complement_graph.number_of_nodes()/3
        graph: networkx.Graph =  networkx.complement(complement_graph)
    else:
        # creation random graph
        if from_erdos:
            n: int = random.randint(min_embeddable, max_embeddable)
            p: float = random.uniform(min_edge_prob, max_edge_prob)
            graph: networkx.Graph = networkx.erdos_renyi_graph(n, p, seed=random_seed, directed=False)
        else:
            check: bool = False
            while check == False:
                name = random.choice(NAME_LIST)
                if TUD_DATA_DICT[name]["data_list"]:
                    index: int = random.choice(TUD_DATA_DICT[name]["data_list"])
                    TUD_DATA_DICT[name]["data_list"].remove(index)
                    check: bool = True
            dataset: TUDataset = TUDataset(root=TUD_DATASET_PATH, name=name, cleaned = True)
            graph: networkx.Graph = torch_geometric.utils.convert.to_networkx(dataset[index], to_undirected=True)
        complement_graph: networkx.Graph = networkx.complement(graph)
    # construction of qubo formulation
    A: int = -1
    B: int = 2
    quadratic: Dict[str, int] = {}
    linear: Dict[str, int] = {(str(i), str(i)):A for i in range(complement_graph.number_of_nodes())}
    for u,v,a in complement_graph.edges(data=True):
      quadratic[(str(u), str(v))]= B
    Q: Dict[str, int] = {**linear, **quadratic}
    # creating input features
    input_graph_features = copy.copy(INPUT_GRAPH_FEATURES_TEMPLATE)
    input_graph_features['Graph_Density'] = networkx.density(complement_graph)
    input_graph_features['Graph_Num_Nodes'] = complement_graph.number_of_nodes()
    input_graph_features['Graph_Num_Edges'] = complement_graph.number_of_edges()
    spectrum=networkx.adjacency_spectrum(complement_graph)
    input_graph_features['Graph_Largest_Eigenvalue'] = abs(spectrum[0]) if len(spectrum)>0 else -1
    input_graph_features['Graph_2nd_Largest_Eigenvalue'] = abs(spectrum[1]) if len(spectrum)>1 else -1
    input_graph_features['Graph_3rd_Largest_Eigenvalue'] = abs(spectrum[2]) if len(spectrum)>2 else -1
    input_graph_features['Graph_4th_Largest_Eigenvalue'] = abs(spectrum[3]) if len(spectrum)>3 else -1
    input_graph_features['Graph_5th_Largest_Eigenvalue'] = abs(spectrum[4]) if len(spectrum)>4 else -1
    largest: int = abs(spectrum[0])
    lowest: int = abs(spectrum[0])
    if input_graph_features['Graph_2nd_Largest_Eigenvalue'] != -1:
        lowest = input_graph_features['Graph_2nd_Largest_Eigenvalue']
    if input_graph_features['Graph_3rd_Largest_Eigenvalue'] != -1:
        lowest = input_graph_features['Graph_3rd_Largest_Eigenvalue']
    if input_graph_features['Graph_4th_Largest_Eigenvalue'] != -1:
        lowest = input_graph_features['Graph_4th_Largest_Eigenvalue']
    if input_graph_features['Graph_5th_Largest_Eigenvalue'] != -1:
        lowest = input_graph_features['Graph_5th_Largest_Eigenvalue']
    input_graph_features['Graph_Spectral_Gap'] = largest-lowest
    tr: int = 0
    triangles = networkx.triangles(complement_graph)
    for label in triangles:
      tr = tr + triangles[label]
    tr = tr/3 # each triangle is counted three times
    input_graph_features['Graph_Num_Triangles'] = tr
    degree_list = [couple[1] for couple in complement_graph.degree]
    input_graph_features['Graph_Min_Degree'] = min(degree_list)
    input_graph_features['Graph_Max_Degree'] = max(degree_list)
    input_graph_features['Graph_Mean_Degree'] = sum(degree_list)/len(degree_list)
    # annealer sampling
    torque_value: float = random.uniform(minimum_torque_value, maximum_torque_value)
    annealing_time: float = random.randint(minimum_anneal_time, maximum_anneal_time)
    qpu_advantage: DWaveSampler = DWaveSampler()
    sampler: EmbeddingComposite = EmbeddingComposite(qpu_advantage)
    alias =  partial(uniform_torque_compensation, prefactor = torque_value) # create alias to compensation
    sampleset = sampler.sample_qubo(Q, num_reads = NUM_READS, return_embedding = True,
                                    auto_scale = True, chain_strength = alias,
                                    annealing_time = annealing_time)
    output_features=copy.copy(OUTPUT_FEATURES_TEMPLATE)
    output_features['qpu_sampling_time'] = sampleset.info['timing']['qpu_sampling_time']
    output_features['qpu_anneal_time_per_sample'] = sampleset.info['timing']['qpu_anneal_time_per_sample']
    output_features['qpu_readout_time_per_sample'] = sampleset.info['timing']['qpu_readout_time_per_sample']
    output_features['qpu_access_time'] = sampleset.info['timing']['qpu_access_time']
    output_features['qpu_access_overhead_time'] = sampleset.info['timing']['qpu_access_overhead_time']
    output_features['qpu_programming_time'] = sampleset.info['timing']['qpu_programming_time']
    output_features['qpu_delay_time_per_sample'] = sampleset.info['timing']['qpu_delay_time_per_sample']
    output_features['total_post_processing_time'] = sampleset.info['timing']['total_post_processing_time']
    output_features['post_processing_overhead_time'] = sampleset.info['timing'][ 'post_processing_overhead_time']
    output_features['Actual_Output_Values'] = []
    for label in sampleset.first[0]:
      if sampleset.first[0][label] == 1:
        output_features['Actual_Output_Values'].append(label)
    output_features['Actual_Output_Values_Len'] = len(output_features['Actual_Output_Values'])
    output_features['Actual_Output_Energy'] = sampleset.first[1]
    output_features['Actual_Output_Occurrences'] = sampleset.first[2]
    output_features['Actual_Output_Chain_Break'] = sampleset.first[3]
    # output of regression and classification
    # note is necessary to check if it is a clique
    H: networkx.Graph = graph.subgraph(output_features['Actual_Output_Values'])
    n: int = len(output_features['Actual_Output_Values'])
    is_clique: bool = False
    if H.size() == n*(n-1)/2:
        output_features['Regression_Output'] =  len(output_features['Actual_Output_Values'])
        is_clique = True
    else:
        output_features['Regression_Output'] = 0
    output_features['Raw_Output'] = len(output_features['Actual_Output_Values'])
    if ((H.size() >= len(networkx.algorithms.approximation.clique.max_clique(graph))) and (is_clique == True)):
        output_features['Classification_Output'] = 1
    else:
        output_features['Classification_Output'] = 0
    # compute autoscale factor
    J, h, b = dimod.utilities.qubo_to_ising(Q)
    max_h_range: int = max(sampler.properties['child_properties']['h_range'])
    min_h_range: int = min(sampler.properties['child_properties']['h_range'])
    max_j_range: int = max(sampler.properties['child_properties']['j_range'])
    min_j_range: int = max(sampler.properties['child_properties']['j_range'])
    factor1: int = max([value for value in h.values()])/max_h_range if h.values() else 0
    factor2: int = min([value for value in h.values()])/min_h_range if h.values() else 0
    factor3: int = max([value for value in J.values()])/max_j_range if J.values() else 0
    factor4: int = min([value for value in J.values()])/min_j_range if J.values() else 0
    autoscale_factor: int = max([max(factor1, 0), max(factor2, 0), max(factor3, 0), max(factor4, 0)])
    if autoscale_factor == 0:
        autoscale_factor = 1
    # create embed graph
    if qpu_advantage.properties['topology']['type'] == 'pegasus':
          working_graph: networkx.Graph = dnx.pegasus_graph(qpu_advantage.properties['topology']['shape'][0],
                                              node_list = qpu_advantage.nodelist,
                                              edge_list = qpu_advantage.edgelist)
    embed_graph: networkx.Graph = networkx.Graph()
    original_edge_list: List[Tuple[Union[int, str]]] = []
    chain_edge_list: List[Tuple[Union[int, str]]] = []
    embed_node_list: List[Tuple[Union[int, str]]] = []
    for label in  sampleset.info['embedding_context']['embedding']:
        chain_len: int = len(sampleset.info['embedding_context']['embedding'][label])
        for i in range(chain_len):
            embed_node_list.append(
                (sampleset.info['embedding_context']['embedding'][label][i],
                {'value': J[str(label)]/autoscale_factor, 'mapping': int(sampleset.info['embedding_context']['embedding'][label][i])}))
            if len(sampleset.info['embedding_context']['embedding'][label]) > 1:
                if ((len(sampleset.info['embedding_context']['embedding'][label]) - (i + 1)) > 0):
                    chain_edge_list.append(
                        (sampleset.info['embedding_context']['embedding'][label][i],
                        sampleset.info['embedding_context']['embedding'][label][i + 1],
                        {'value': sampleset.info['embedding_context']['chain_strength'],
                        'mapping_1': int(sampleset.info['embedding_context']['embedding'][label][i]),
                        'mapping_2': int(sampleset.info['embedding_context']['embedding'][label][i+1])}))
    for couple in h:
        first: Tuple[Union[int, Dict[str, int]]] = couple[0]
        second: Tuple[Union[int, Dict[str, int]]]  = couple[1]
        if ((len(sampleset.info['embedding_context']['embedding'][first]) == 1) and (len(sampleset.info['embedding_context']['embedding'][second]) == 1)):
            original_edge_list.append(( sampleset.info['embedding_context']['embedding'][first][0],
                                      sampleset.info['embedding_context']['embedding'][second][0],
                                        {'value': h[couple]/autoscale_factor,
                                        'mapping_1': int(sampleset.info['embedding_context']['embedding'][first][0]),
                                        'mapping_2': int(sampleset.info['embedding_context']['embedding'][second][0])}))
        elif ((len(sampleset.info['embedding_context']['embedding'][first]) != 1) and (len(sampleset.info['embedding_context']['embedding'][second]) == 1)):
            chain_len: int = len(sampleset.info['embedding_context']['embedding'][first])
            found: bool = False
            for i in range(chain_len):
                if found == False:
                    if working_graph.has_edge(sampleset.info['embedding_context']['embedding'][first][i],
                                              sampleset.info['embedding_context']['embedding'][second][0]):
                          original_edge_list.append(( sampleset.info['embedding_context']['embedding'][first][i],
                                                    sampleset.info['embedding_context']['embedding'][second][0],
                                                      {'value': h[couple]/autoscale_factor,
                                                      'mapping_1':int(sampleset.info['embedding_context']['embedding'][first][i]),
                                                      'mapping_2': int(sampleset.info['embedding_context']['embedding'][second][0])}))
                          found = True
        elif ((len(sampleset.info['embedding_context']['embedding'][first]) == 1) and (len(sampleset.info['embedding_context']['embedding'][second]) != 1)):
            chain_len: int = len(sampleset.info['embedding_context']['embedding'][second])
            found: bool = False
            for i in range(chain_len):
                if found == False:
                    if working_graph.has_edge(sampleset.info['embedding_context']['embedding'][first][0],
                                              sampleset.info['embedding_context']['embedding'][second][i]):
                        original_edge_list.append(( sampleset.info['embedding_context']['embedding'][first][0],
                                                  sampleset.info['embedding_context']['embedding'][second][i],
                                                    {'value': h[couple]/autoscale_factor,
                                                    'mapping_1': int(sampleset.info['embedding_context']['embedding'][first][0]),
                                                    'mapping_2': int(sampleset.info['embedding_context']['embedding'][second][i])}))
                        found: bool = True
        else:
            chain_len_first: int = len(sampleset.info['embedding_context']['embedding'][first])
            chain_len_second: int = len(sampleset.info['embedding_context']['embedding'][second])
            found: bool = False
            for i in range(chain_len_first):
                for j in range(chain_len_second):
                    if working_graph.has_edge(sampleset.info['embedding_context']['embedding'][first][i],
                                              sampleset.info['embedding_context']['embedding'][second][j]):
                          original_edge_list.append(( sampleset.info['embedding_context']['embedding'][first][i],
                                                    sampleset.info['embedding_context']['embedding'][second][j],
                                                      {'value': h[couple]/autoscale_factor,
                                                      'mapping_1': int(sampleset.info['embedding_context']['embedding'][first][i]),
                                                      'mapping_2': int(sampleset.info['embedding_context']['embedding'][second][j])}))
                    found: bool = True
    embed_graph.add_nodes_from(embed_node_list)
    embed_graph.add_edges_from(original_edge_list)
    embed_graph.add_edges_from(chain_edge_list)
    # creating embed graph features
    embed_graph_features: Dict[str. Union[int, float]] = copy.copy(EMBED_GRAPH_FEATURES_TEMPLATE)
    embed_graph_features['Graph_Density'] = networkx.density(embed_graph)
    embed_graph_features['Graph_Num_Nodes'] = embed_graph.number_of_nodes()
    embed_graph_features['Graph_Num_Edges'] = embed_graph.number_of_edges()
    spectrum: int = networkx.adjacency_spectrum(embed_graph)
    embed_graph_features['Graph_Largest_Eigenvalue'] = abs(spectrum[0]) if len(spectrum)>0 else -1
    embed_graph_features['Graph_2nd_Largest_Eigenvalue'] = abs(spectrum[1]) if len(spectrum)>1 else -1
    embed_graph_features['Graph_3rd_Largest_Eigenvalue'] = abs(spectrum[2]) if len(spectrum)>2 else -1
    embed_graph_features['Graph_4th_Largest_Eigenvalue'] = abs(spectrum[3]) if len(spectrum)>3 else -1
    embed_graph_features['Graph_5th_Largest_Eigenvalue'] = abs(spectrum[4]) if len(spectrum)>4 else -1
    largest: int = abs(spectrum[0])
    lowest: int = abs(spectrum[0])
    if embed_graph_features['Graph_2nd_Largest_Eigenvalue'] != -1:
        lowest = embed_graph_features['Graph_2nd_Largest_Eigenvalue']
    if embed_graph_features['Graph_3rd_Largest_Eigenvalue'] != -1:
        lowest = embed_graph_features['Graph_3rd_Largest_Eigenvalue']
    if embed_graph_features['Graph_4th_Largest_Eigenvalue'] != -1:
        lowest = embed_graph_features['Graph_4th_Largest_Eigenvalue']
    if embed_graph_features['Graph_5th_Largest_Eigenvalue'] != -1:
        lowest = embed_graph_features['Graph_5th_Largest_Eigenvalue']
    embed_graph_features['Graph_Spectral_Gap'] = largest-lowest
    tr: int = 0
    triangles = networkx.triangles(embed_graph)
    for label in triangles:
        tr = tr + triangles[label]
    tr = tr/3 # each triangle is counted three times
    embed_graph_features['Graph_Num_Triangles']=tr
    degree_list: List[Tuple[int]] = [couple[1] for couple in embed_graph.degree]
    embed_graph_features['Graph_Min_Degree'] = min(degree_list)
    embed_graph_features['Graph_Max_Degree'] = max(degree_list)
    embed_graph_features['Graph_Mean_Degree'] = sum(degree_list)/len(degree_list)
    chain_lengths: List[int] = []
    for label in  sampleset.info['embedding_context']['embedding']:
        chain_lengths.append(len(sampleset.info['embedding_context']['embedding'][label]))
    embed_graph_features['Min_Chain_Length'] = min(chain_lengths)
    embed_graph_features['Max_Chain_Length'] = max(chain_lengths)
    embed_graph_features['Avg_Chain_Length'] = sum(chain_lengths)/len(chain_lengths)
    embed_graph_features['Chain_Strength'] = sampleset.info['embedding_context']['chain_strength']
    embed_graph_features['Annealing_Time'] = annealing_time
    if embed_graph.edges(data=True):
        pyg_graph: torch_geometric.data.Data = torch_geometric.utils.convert.from_networkx(
            embed_graph, ['value', 'mapping'], ['value', 'mapping_1', 'mapping_2'])
    else:
         pyg_graph: torch_geometric.data.Data = torch_geometric.utils.convert.from_networkx(
             embed_graph, ['value', 'mapping'])
    if pyg_graph.edge_attr is None:
        pyg_graph.edge_attr = torch.full((pyg_graph.edge_index.shape[1], 3), -1)
    pyg_graph.y = torch.tensor(output_features["Classification_Output"].item())
    pyg_graph.y_regr = torch.tensor(output_features["Regression_Output"].item())
    feats = [input_graph_features[lab] for lab in input_graph_features]
    feats = feats.extend([embed_graph_features[lab] for lab in embed_graph_features])
    pyg_graph.features = torch.tensor(feats)
    # add information of input graph
    for u,v,a in complement_graph.edges(data=True):
      networkx.set_edge_attributes(complement_graph,{(u, v):2}, name = "value")
    for i in range(complement_graph.number_of_nodes()):
        networkx.set_node_attributes(complement_graph,  {i:-1}, name = "value")
    if complement_graph.edges(data=True):
        pyg_graph_input = torch_geometric.utils.convert.from_networkx(
            complement_graph, ['value'], ['value'])
    else:
        pyg_graph_input = torch_geometric.utils.convert.from_networkx(
            complement_graph, ['value'])
    if pyg_graph_input.edge_attr is None:
        pyg_graph_input.edge_attr = torch.zeros(pyg_graph_input.edge_index.shape[1], 1)
    pyg_graph.input_edge_index = pyg_graph_input.edge_index
    pyg_graph.input_x = pyg_graph_input.x
    pyg_graph.input_edge_attr = pyg_graph_input.edge_attr
    data: Dict[str, Union[int, float]] = pyg_graph.to_dict()
    return input_graph_features, embed_graph_features, output_features, data

```

#####**5.1.1 Maximum clique**

maximum clique [1] graph problems are typical benchmarks. Here this is selected for comparison with barbosa. Other probles like weighted max cut [2] are directly embeddable in Ising hamiltonian similarly to the task at hand.
The maximum clique problem is equivalent to the maximum independent set problem and the vertex cover problem. It is also used as an intermediate step in solving the maximum common subgraph problem. Here it is considered an optimization version meaning we are trying finding greatest clique possible


Given an undirected graph G = (V, E), a clique is
a subset S of the vertices forming a complete subgraph, meaning that any two vertices of S are
connected by an edge in G. The clique size is the number of vertices in S, and the maximum clique
problem is to find a clique with a maximum number of vertices in G [3]

---



ising formulation is taken from [4]

Recall that a QUBO problem can be written as
$$
\underset{x_i \in\{0,1\}}{\operatorname{minimize}} H=\sum_{1 \leq i<j \leq N} a_{i j} x_i x_j,
$$
where the weights $a_{i j}, i \neq j$, are the quadratic terms and $a_{i i}$ are the linear terms (since $x_i^2=x_i$ for $\left.x_i \in\{0,1\}\right)$.

There are multiple ways to formulate the MC problem as a QUBO. One of the simplest is based on the equivalence between $\mathrm{MC}$ and the maximum independent set problem. An independent set $S$ of a graph $H$ is a set of vertices with the property that for any two vertices $v, w \in S, v$ and $w$ are not connected by an edge in $H$. It is easy to see that an independent set of $H=(V, \bar{E})$ defines a clique in graph $G=(V, E)$, where $\bar{E}$ is the complement of set $E$. Therefore, looking for the maximum clique in $G$ is equivalent to finding the maximum independent set in $H$. The corresponding constraint formulation for $\mathrm{MC}$ is
$$
\begin{array}{ll}
\underset{x_i \in\{0,1\}}{\operatorname{maximize}} & \sum_{i=1}^N x_i \\
\text { subject to } & \sum_{(i, j) \in \bar{E}} x_i x_j=0,
\end{array}
$$
where $G=(V, E)$ is the input graph and $\bar{E}$ is the complement of $E$. The equivalent unconstrained (QUBO) minimization of previous equations, written in the form of first equation, is
$$
H=-A \sum_{i=1}^N x_i+B \sum_{(i, j) \in \bar{E}} x_i x_j,
$$
where one can determine that the coefficients/penalties $A$ and $B$ can be chosen as $A=1, B=2$ [5]

Auto-scaling works as follows. Each QPU has an allowed range of values for the biases and strengths of qubits and couplers. Unless you explicitly disable auto-scaling, the values defined in your problem are adjusted to fit the entire available range, by dividing them by a positive (non-zero) factor defined as:
$$
\begin{array}{r}
\max \left\{\max \left(\frac{\max (h)}{\max \left(h \_ \text {range }\right)}, 0\right),\right. \\
\max \left(\frac{\min (h)}{\min \left(h \_ \text {range }\right)}, 0\right), \\
\max \left(\frac{\max (J)}{\max \left(J \_ \text {range }\right)}, 0\right), \\
\left.\max \left(\frac{\min (J)}{\min \left(J \_ \text {range }\right)}, 0\right)\right\}
\end{array}

#####**5.1.2 3-SAT**

max sat [Reinforcement Quantum Annealing: A Hybrid Quantum Learning Automata]

a lot of problems can be converted to sat . e.g.  pseudo prime numbers [Reinforcement Quantum Annealing: A Hybrid Quantum Learning Automata] espongono modo per trasformarla in SAT che è uno dei problemi NP piu noti e per cui piu riduzioni polinomiali sono note. inoltre viene considerato in molti lavori sull annealer (e.g. [Reinforcement Quantum Annealing: A Hybrid Quantum Learning Automata] )

here is considered a decision version

per la formulazione utilizziamo [Farhi, E., Goldstone, J., Gutmann, S., Sipser, M.: Quantum computation by adiabatic evolution. arXiv preprint quant-ph/0001106 (2000)][Choi, V.: Adiabatic quantum algorithms for the NP-complete Maximum-Weight
Independent set, Exact Cover and 3SAT problems. arXiv:1004.2226 (2010)][Choi, V.: Different adiabatic quantum optimization algorithms for the NP-complete
exact cover and 3SAT problems. Quant. Inform. & Comp. 11(7-8) (2011)]

Similarly, for 3SAT, there is a well-known reduction to MIS (which is one of the first NP-complete reductions, to show the NP-hardness of MIS) [M. R. Garey and D. S. Johnson (1979), Computers and Intractability: A Guide to the Theory of NPCompleteness, W. H. Freeman.]. For completeness, here we recall the reduction:
3SAT $\leq_P$ MIS. Given a 3SAT instance $\Psi\left(x_1, \ldots, x_n\right)=C_1 \wedge \ldots \wedge C_m$ with $n$ variables and $m$ clauses, we construct a (unweighted) graph $G_{\text {SAT }}$ as follows:
- For each clause $C_i=y_{i_1} \vee y_{i_2} \vee y_{i_3}$, we construct a triangle with three vertices labeled accordingly, i.e., with $y_{i_1}, y_{i_2}, y_{i_3}$, where $y_j \in\left\{x_j, \bar{x}_j\right\}$. Therefore, $G_{\mathrm{SAT}}$ consists of $3 m$ vertices.
- There is an edge between two vertices in different triangles if there labels are in conflict. That is, for $i \neq j$, $i_s j_t \in \mathrm{E}\left(G_{\mathrm{SAT}}\right)$ if and only if $y_{i_s}=\bar{y}_{j_t}$.

One can then show that $\Psi$ is satisfiable if and only if $G_{\text {SAT }}$ has a MIS of size $m$.

```python
def construct_autoencoder_data(num: int = 1000, num_variation: int = 5, num_ops: int = 10) -> None:
    '''
    construct autoencoder dataset
    '''
    # retrieve all data
    # clique erdos
    erdos_list_dict = None
    with open(DATA_PATH_CLIQUE_ERDOS, 'rb') as f:
                  erdos_list_dict = pickle.load(f)
    full_dict_list: List[Dict[str, Union[float, int]]] = copy.deepcopy(erdos_list_dict)
    full_data_unpolished: List[torch_geometric.data.Data] = []
    for dic in full_dict_list:
        new_data: torch_geometric.data.Data = Data().from_dict(dic)
        full_data_unpolished.append(new_data)
    full_data = []
    for data in full_data_unpolished: # remove data without edges
        if data.edge_attr is not None:
            full_data.append(data)
    # list where store data
    data_list_real, data_list_fake = [], []  # store data generated
    label_list: List[int] = []  # store label generated
    for i in range(num):  # construct df iteratively
        print("iteration " + str(i))
        # select random graph
        reference_graph: networkx.Graph = random.choice(full_data)
        # set to none some element
        reference_graph_nx = torch_geometric.utils.convert.to_networkx(
            reference_graph, to_undirected= True, edge_attrs = ["edge_attr"],
            node_attrs = ["x"]) # nx version of the reference graph
        # append number of variation time
        data_list_real.extend([reference_graph for i in range(num_variation)])
        # iterate number variation
        for j in range(num_variation):
            max_strength: int = max([torch.max(
                reference_graph.x[:, 0]).item(),
                torch.max(reference_graph.x[:, 0]).item()]) # maximum strength inside graph
            index_list: List[int] = reference_graph.x[:, 1].tolist()
            index_set: Set[int] = set( index_list) # index set
            edge_set: Set[Tuple[int]] = set(list(reference_graph_nx.edges()))
            index_dict: Dict[int, List[int]] = {i:index_list[i] for i in range(len(index_list))}
            max_index: int = max(index_set) # maximum index inside graph
            # want always at least a valid variation
            all_valid: bool = False
            if (j < int(num_variation/2)): # this is done to obtain balanced dataset
                all_valid = True
            # do random number of operation
            label: int = 1
            extracted_num_op: int = random.randrange(1, num_ops + 1)
            altered_graph: networkx.Graph = copy.deepcopy(reference_graph_nx)
            for k in range(extracted_num_op):
                choose: int = random.randrange(0, 3) if not(all_valid) else random.randrange(0, 2) # only choose third option if not all valid
                # add a node
                if choose == 0:
                    new_index: int = random.randrange(max_index, 2*max_index)
                    altered_graph.add_node(altered_graph.number_of_nodes(), x = [1.0, new_index])
                    correct: int = random.randrange(0, 2) if not(all_valid) else 0  # only not correct if not all valid
                    if correct == 0: # correct adding an edge (else invalid)
                        first: int = altered_graph.number_of_nodes()-1
                        second: int = random.choice(list(index_dict.keys()))
                        while ((second == first) and ((first, second) not in edge_set)):
                            second = random.choice(list(index_dict.keys()))
                        altered_graph.add_edge(first, second, edge_attr=[
                            random.uniform(max_strength, 2*max_strength), new_index, index_dict[second]])
                    else: # always uncorrect
                        label: int = 0
                # multiply node and/or edge values
                if choose == 1:
                    what_multiply: int = random.randrange(0, 3)  if not(all_valid) else 0
                    multiplier: float = random.uniform(0, 2*max_strength)
                    if what_multiply == 0: #  multiply both (always valid)
                        for couple in list(edge_set):
                          altered_graph.edges[couple]["edge_attr"][0] = multiplier*altered_graph.edges[couple]["edge_attr"][0]
                        for i in range(len(index_list)):
                          altered_graph.nodes[i]["x"][0] = multiplier*altered_graph.nodes[i]["x"][0]
                    elif what_multiply == 1: #  multiply node (always invalid)
                        for i in range(len(index_list)):
                            altered_graph.nodes[i]["x"][0] = multiplier*altered_graph.nodes[i]["x"][0]
                        label = 0
                    elif what_multiply == 2: #  multiply edge (always invalid)
                        for couple in list(edge_set):
                            altered_graph.edges[couple]["edge_attr"][0] = multiplier*altered_graph.edges[couple]["edge_attr"][0]
                        label = 0
                # add a vertex (always invalid op)
                if choose == 2:
                    first: int = random.choice(list(index_dict.keys()))
                    second: int = random.choice(list(index_dict.keys()))
                    while ((second == first) and ((first, second) not in edge_set)):
                        second = random.choice(list(index_dict.keys()))
                    altered_graph.add_edge(first, second, edge_attr=[
                        random.uniform(0, 2*max_strength), index_dict[first], index_dict[second]])
                    label: int = 0 # always uncorrect
                index_list: List[int] = [altered_graph.nodes[i]["x"][1] for i in range(altered_graph.number_of_nodes())]
                index_set: Set[int] = set(index_list) # index set
                edge_set: Set[Tuple[int]] = set(list(altered_graph.edges()))
                index_dict: Dict[int, List[int]] = {i:index_list[i] for i in range(len(index_list))}
                max_index: int = max(index_set) # maximum index inside graph
            data_list_fake.append(torch_geometric.utils.convert.from_networkx(altered_graph, ['x'], ['edge_attr']))
            label_list.append(label)
    # create autoencoder format
    data_list_real: List[torch_geometric.data.Data] = [train_test_split_edges(copy.deepcopy(data_list_real[i])) for i in range(len(data_list_real))]
    data_list_fake: List[torch_geometric.data.Data] = [train_test_split_edges(copy.deepcopy(data_list_fake[i])) for i in range(len(data_list_fake))]
    # avoid no edges
    temp_list_real: List[torch_geometric.data.Data] = []
    temp_list_fake: List[torch_geometric.data.Data] = []
    temp_list_label: List[torch_geometric.data.Data] = []
    i = 0
    for data in data_list_real:
        if not((data.val_pos_edge_index.shape[1] == 0) or (data.val_neg_edge_index.shape[1] == 0)
                or (data.test_pos_edge_index.shape[1] == 0) or (data.test_neg_edge_index.shape[1] == 0)):
          if not((data_list_fake[i].val_pos_edge_index.shape[1] == 0) or (data_list_fake[i].val_neg_edge_index.shape[1] == 0)
                  or (data_list_fake[i].test_pos_edge_index.shape[1] == 0) or (data_list_fake[i].test_neg_edge_index.shape[1] == 0)):
              temp_list_real.append(data)
              temp_list_fake.append(data_list_fake[i])
              temp_list_label.append(AUTOECONDER_LABELS[i])
        i = i + 1
    data_list_real = temp_list_real
    data_list_fake = temp_list_fake
    label_list = temp_list_label
    # real data
    with open(DATA_PATH_REAL, 'wb') as f:
      pickle.dump(data_list_real, f)
    # fake data
    with open(DATA_PATH_FAKE, 'wb') as f:
      pickle.dump(data_list_fake, f)
    # label
    with open(LABEL_PATH, 'wb') as f:
      pickle.dump(label_list, f)
    return None
```

```python
def get_reinforce_data(maximum_node: int = 50, maximum_edge: int = 50) -> List[Dict[str, Union[float, int]]]:
    '''
    construct reinforcement learning data
    '''
    # real clique data
    real_list_dict: List[Dict[str, Union[float, int]]] = None
    with open(DATA_PATH_CLIQUE_TUD, 'rb') as f:
                  real_list_dict = pickle.load(f)
    real_data = [Data().from_dict(real_list_dict[i]) for i in range(len(real_list_dict))]
    returned_list: List[Dict[str, Union[float, int]]] = []
    for data in real_data:
        if data.y == 0:
              data_reinforcement = copy.deepcopy(DATA_REINFORCMENT_TEMPLATE)
              input_graph: torch_geometric.data.Data = Data()
              input_graph.x = data.input_x
              input_graph.edge_index = data.input_edge_index
              input_graph.edge_attr = data.input_edge_attr
              data_reinforcement["Input_Graph"] = input_graph.to_dict()
              complement_graph: networkx.Graph = torch_geometric.utils.convert.from_networkx(input_graph, group_node_attrs=['value', 'mapping'], group_edge_attrs = ["value", "mapping_1", "mapping_2"])
              graph: networkx.Graph =  networkx.complement(complement_graph)
              data_reinforcement["Input_Label"] = len(networkx.algorithms.approximation.clique.max_clique(graph))
              # creating input feaures
              input_graph_features: Dict[str, Union[int, float]] = copy.copy(INPUT_GRAPH_FEATURES_TEMPLATE)
              input_graph_features['Graph_Density']=networkx.density(complement_graph)
              input_graph_features['Graph_Num_Nodes']=complement_graph.number_of_nodes()
              input_graph_features['Graph_Num_Edges']=complement_graph.number_of_edges()
              spectrum: float = networkx.adjacency_spectrum(complement_graph)
              input_graph_features['Graph_Largest_Eigenvalue']=abs(spectrum[0]) if len(spectrum)>0 else -1
              input_graph_features['Graph_2nd_Largest_Eigenvalue']=abs(spectrum[1]) if len(spectrum)>1 else -1
              input_graph_features['Graph_3rd_Largest_Eigenvalue']=abs(spectrum[2]) if len(spectrum)>2 else -1
              input_graph_features['Graph_4th_Largest_Eigenvalue']=abs(spectrum[3]) if len(spectrum)>3 else -1
              input_graph_features['Graph_5th_Largest_Eigenvalue']=abs(spectrum[4]) if len(spectrum)>4 else -1
              largest: int = abs(spectrum[0])
              lowest: int = abs(spectrum[0])
              if input_graph_features['Graph_2nd_Largest_Eigenvalue'] != -1:
                  lowest=input_graph_features['Graph_2nd_Largest_Eigenvalue']
              if input_graph_features['Graph_3rd_Largest_Eigenvalue'] != -1:
                  lowest=input_graph_features['Graph_3rd_Largest_Eigenvalue']
              if input_graph_features['Graph_4th_Largest_Eigenvalue'] != -1:
                  lowest=input_graph_features['Graph_4th_Largest_Eigenvalue']
              if input_graph_features['Graph_5th_Largest_Eigenvalue'] != -1:
                  lowest=input_graph_features['Graph_5th_Largest_Eigenvalue']
              input_graph_features['Graph_Spectral_Gap']=largest-lowest
              tr: int = 0
              triangles = networkx.triangles(complement_graph)
              for label in triangles:
                tr = tr + triangles[label]
              tr = tr/3 # each triangle is counted three times
              input_graph_features['Graph_Num_Triangles'] = tr
              degree_list: List[Tuple[int]] = [couple[1] for couple in complement_graph.degree]
              input_graph_features['Graph_Min_Degree'] = min(degree_list)
              input_graph_features['Graph_Max_Degree']= max(degree_list)
              input_graph_features['Graph_Mean_Degree'] = sum(degree_list)/len(degree_list)
              data_reinforcement["Input_Graph_Features"] = input_graph_features
              if ((len(networkx.algorithms.approximation.clique.max_clique(graph)) > 0)
                  and (complement_graph.number_of_nodes() < 100)
                  and (complement_graph.number_of_edges() < 100)):
                  returned_list.append(data_reinforcement)
    return returned_list
```

```python
def oversample_data(data: List[torch_geometric.data.Data], labels: np.ndarray) -> List[Union[List[torch_geometric.data.Data] , np.ndarray]]:
    '''
    oversample dataset
    '''
    data: List[torch_geometric.data.Data] = copy.deepcopy(data)
    labels: List[int] = copy.deepcopy(labels).tolist()
    frequencies: Dict[int, int] = {}
    i: int = 0
    for lab in labels:
        if int(lab) in frequencies:
            frequencies[int(lab)]["count"] += 1
            frequencies[int(lab)]["store"].append(data[i])
        else:
            frequencies[int(lab)] = {}
            frequencies[int(lab)]["count"] = 1
            frequencies[int(lab)]["store"] = [data[i]]
        i = i + 1
    max_freq: int = max([frequencies[int(lab)]["count"] for lab in frequencies])
    for lab in frequencies:
        if max_freq != frequencies[lab]["count"]:
            for i in range(max_freq -  frequencies[int(lab)]["count"]):
                data_item: torch_geometric.data.Data = random.choice(frequencies[int(lab)]["store"])
                data.append(data_item)
                labels.append(data_item.y)
    return data, np.array(labels)
```

```python
class MessageBlock(torch.nn.Module):
  '''
  a message convolutional block composed by a convolution followed by normalization followed by dropout
  '''
  def __init__(self, config: Dict[str, Union[str, int, float]] = CONFIG_TEMPLATE_PREDICTOR, i: int = 0) -> None:
      super(MessageBlock, self).__init__()
      self.config: Dict[str, Union[str, int, float]] = config
      self.num_node_features: int = FULL_DATA[0].num_node_features
      self.num_edge_features: int = FULL_DATA[0].num_edge_features
      self.hidden_channels: int = config["net"]["hidden_channels"]
      # activation
      if self.config["net"]["activation"] == "relu":
          self.activation: torch.nn.ReLU = torch.nn.ReLU()
      elif self.config["net"]["activation"] == "prelu":
          self.activation: torch.nn.PReLU = torch.nn.PReLU(num_parameters=self.config["net"]["prelu_num_parameters"], init=self.config["net"]["prelu_init"])
      elif self.config["net"]["activation"] == "swish":
        self.activation: torch.nn.SiLU = torch.nn.SiLU()
      # convolution
      self.conv = None
      if self.config["net"]["type_conv"] == "gat":
          self.conv: GATConv = GATConv(self.num_node_features if i == 0 else self.hidden_channels, self.hidden_channels, edge_dim = self.num_edge_features,
                              negative_slope = self.config["net"]["negative_slope"]) # ignore dropout here given we apply it thereafter
      elif self.config["net"]["type_conv"] == "reset":
          self.conv: ResGatedGraphConv = ResGatedGraphConv(self.num_node_features if i == 0 else self.hidden_channels, self.hidden_channels, edge_dim = self.num_edge_features,
                              act = self.activation) # pass same activation
      elif self.config["net"]["type_conv"] == "conv":
          self.conv: GATv2Conv = GATv2Conv(self.num_node_features if i == 0 else self.hidden_channels, self.hidden_channels, edge_dim = self.num_edge_features,
                                negative_slope = self.config["net"]["negative_slope"]) # pass same activation
      elif self.config["net"]["type_conv"] == "pdn":
          self.conv: PDNConv = PDNConv(self.num_node_features if i == 0 else self.hidden_channels,
                              self.hidden_channels, edge_dim = self.num_edge_features,
                              hidden_channels = self.hidden_channels )
      # normalization
      if self.config["net"]["normalization"] is not None:
          if self.config["net"]["normalization"] == "batch":
              self.norm: BatchNorm = BatchNorm(config["net"]["hidden_channels"] )
          elif self.config["net"]["normalization"] == "graph":
              self.norm: GraphNorm = GraphNorm(config["net"]["hidden_channels"] )
          if self.config["net"]["normalization"] == "pair":
              self.norm: PairNorm = PairNorm(config["net"]["hidden_channels"] )
          if self.config["net"]["normalization"] == "mean":
              self.norm: MeanSubtractionNorm = MeanSubtractionNorm( )
      return None

  def forward(self, x: torch.tensor, data: torch_geometric.data.Data) -> torch.tensor:
      # message
      x: torch.tensor = self.conv(x, data.edge_index, edge_attr = data.edge_attr)
      # batch norm
      if self.config["net"]["normalization"] is not None:
          x = self.norm(x)
      # dropout
      x = F.dropout(x, p=self.config["net"]["dropout"], training=self.training)
      # activation
      x = self.activation(x)
      return x
```

```python
class PredictorNet(torch.nn.Module):
    '''
    predictor neural network composed bylinear preproces layer folowed by convolutional block
    as defined above folowed by postporcess layer
    '''
    def __init__(self, task: str = "classification", config: Dict[str, Union[str, int, float]]  = CONFIG_TEMPLATE_PREDICTOR) -> None:
        super(PredictorNet, self).__init__()
        self.task: str = task
        self.config: Dict[str, Union[str, int, float]]  = config
        self.num_node_features: int = FULL_DATA[0].num_node_features
        self.num_edge_features: int = FULL_DATA[0].num_edge_features
        self.hidden_channels: int = config["net"]["hidden_channels"]
        self.computed_features: int = FULL_DATA[0].features.shape[0]
        # preprocess layer
        self.preprocess_layer: nn.ModuleList = None
        if self.config["net"]["n_preprocess_layer"] != 0:
            self.preprocess_layer = nn.ModuleList([nn.Linear(self.num_node_features, self.num_node_features) for i in range(self.config["net"]["n_preprocess_layer"])])
        # message passing layer
        self.messag_passing_layers: nn.ModuleList = None
        if self.config["net"]["n_message_passing_layer"] != 0:
            self.messag_passing_layers = nn.ModuleList([MessageBlock(config, i) for i in range(self.config["net"]["n_message_passing_layer"])])
        # post process layer
        self.postprocess_layer: nn.ModuleList = None
        output: int = 2 if self.task == "classification" else 1
        if self.config["net"]["n_postprocess_layer"] != 0:
            self.postprocess_layer = nn.ModuleList()
            if self.config["net"]["jumping_knowledge"] is None:
                  input = self.hidden_channels + self.computed_features
            else:
                  input = self.config["net"]["n_message_passing_layer"]*self.hidden_channels + self.computed_features
            for i in range(self.config["net"]["n_postprocess_layer"]):
                  self.postprocess_layer.append(nn.Linear(input, output if i == self.config["net"]["n_postprocess_layer"] - 1 else int(input/2)))
                  input = int(input/2)
        # jumping knowledge
        if self.config["net"]["jumping_knowledge"] is not None:
            self.jk: JumpingKnowledge = JumpingKnowledge(self.config["net"]["jumping_knowledge"], self.hidden_channels, self.config["net"]["n_message_passing_layer"])
        # activation
        if self.config["net"]["activation"] == "relu":
            self.activation: torch.nn.ReLU = torch.nn.ReLU()
        elif self.config["net"]["activation"] == "prelu":
            self.activation: torch.nn.PReLU = torch.nn.PReLU(num_parameters=self.config["net"]["prelu_num_parameters"], init=self.config["net"]["prelu_init"])
        elif self.config["net"]["activation"] == "swish":
            self.activation: torch.nn.SiLU = torch.nn.SiLU()
        return None

    def forward(self, data: torch_geometric.data.Data) -> torch.tensor:
        x: torch.tensor = data.x
        # Apply preprocess layer
        for layer in self.preprocess_layer:
            x = layer(x)
            x = self.activation(x)
        # Apply message passing blocks
        xs: List[torch.tensor] = []
        i: int = 0
        for layer in self.messag_passing_layers:
            pre_x = x.clone()
            x = layer(x, data)
            xs.append(x)
            i = i + 1
        # apply jumping knowledge
        if self.config["net"]["jumping_knowledge"] is not None:
           x = self.jk(xs)
        # apply global mean
        if self.config["net"]["aggregation"] == "mean":
            x = global_mean_pool(x, data.batch)
        elif self.config["net"]["aggregation"] == "add":
            x = global_add_pool(x, data.batch)
        elif self.config["net"]["aggregation"] == "max":
            x = global_max_pool(x, data.batch)
        reshaped_features = torch.reshape(data.features, (data.ptr.shape[0] -1, self.computed_features))
        reshaped_features = reshaped_features.type(torch.FloatTensor)
        x = torch.cat([x, reshaped_features.to(DEVICE)], dim=1)
        # Apply post processing layers
        i = 0
        for layer in self.postprocess_layer:
            x = layer(x)
            if i != self.config["net"]["n_postprocess_layer"] - 1:
                x = self.activation(x)
        return x
```

Teorem 1. For any problem in class NP, there are infnite diferent Ising models whose ground states are all identical to the solution of the original problem.

Proof. According to Cook—Levin theorem, we can reduce any NP problem to finding the ground state of Hamiltonians (which is also in the class NP) in polynomial-time. Multiplying all coefcients of the Ising model
by a positive non-zero real number will result in a new spin glass problem whose ground state will be identical to the original Ising model. Since the number of positive real numbers are infnite, we can generate infnite diferent
Ising models whose ground states represent the solution for the original problem of interest

```python
class Encoder(torch.nn.Module):
    '''
    encoder class
    '''
    def __init__(self,  config: Dict[str, Union[float, int, str]] = CONFIG_TEMPLATE_AUTOENCODER) -> None:
        super(Encoder, self).__init__()
        self.config: Dict[str, Union[float, int, str]] = config
        self.num_edge_features: int = AUTOENCODER_REAL_DATA[0].num_edge_features
        self.in_channels: int = AUTOENCODER_REAL_DATA[0].num_features
        self.hidden_channels: int =  config["net"]["hidden_channels"]
        self.out_channels: int = config["net"]["out_channels"]
         # activation
        if self.config["net"]["activation"] == "relu":
            self.activation: torch.nn.ReLU = torch.nn.ReLU()
        elif self.config["net"]["activation"] == "prelu":
            self.activation: torch.nn.PReLU = torch.nn.PReLU(
                num_parameters = self.config["net"]["prelu_num_parameters"],
                init = self.config["net"]["prelu_init"])
        elif self.config["net"]["activation"] == "swish":
          self.activation: torch.nn.SiLU = torch.nn.SiLU()
        # preprocessing
        self.preprocess_layer: nn.ModuleList = None
        if self.config["net"]["n_preprocess_layer"] != 0:
            self.preprocess_layer = nn.ModuleList([nn.Linear(self.in_channels, self.in_channels) for i in range(self.config["net"]["n_preprocess_layer"])])
        if self.config["net"]["n_message_passing_layer"] != 0:
            input: int = self.in_channels
            output: int = self.hidden_channels
            self.conv_layers: nn.ModuleList = nn.ModuleList()
            self.norms: nn.ModuleList = nn.ModuleList()
            for i in range(self.config["net"]["n_message_passing_layer"] ):
                # convolution
                if ((i != 0) and (i != self.config["net"]["n_message_passing_layer"] - 1)):
                    input = output
                    output = int(output/2) if int(output/2) > self.out_channels else self.out_channels
                elif ((i != 0) and (i == self.config["net"]["n_message_passing_layer"] - 1)):
                    input = output
                    output = self.out_channels
                if self.config["net"]["type_conv"] == "gat":
                    self.conv_layers.append(GATConv(input, output, edge_dim = self.num_edge_features,
                                            negative_slope = self.config["net"]["negative_slope"])) # ignore dropout here given we apply it thereafter
                elif self.config["net"]["type_conv"] == "reset":
                    self.conv_layers.append(ResGatedGraphConv(input, output, edge_dim = self.num_edge_features,
                                        act = self.activation)) # pass same activation
                elif self.config["net"]["type_conv"] == "conv":
                    self.conv_layers.append(GATv2Conv(input, output, edge_dim = self.num_edge_features,
                                           negative_slope = self.config["net"]["negative_slope"])) # pass same activation
                elif self.config["net"]["type_conv"] == "pdn":
                    self.conv_layers.append(PDNConv(input, output, edge_dim = self.num_edge_features))
                # normalization
                if self.config["net"]["normalization"] is not None:
                    if self.config["net"]["normalization"] == "batch":
                        self.norms.append(BatchNorm(output))
                    elif self.config["net"]["normalization"] == "graph":
                        self.norms.append(GraphNorm(output))
                    elif self.config["net"]["normalization"] == "pair":
                        self.norms.append(PairNorm(output))
                    elif self.config["net"]["normalization"] == "mean":
                        self.norms.append(MeanSubtractionNorm())
        return None

    def forward(self, data: torch_geometric.data.Data) -> torch.tensor:
        x: torch.tensor = data.x
        # Apply preprocess layer
        if self.config["net"]["n_preprocess_layer"] != 0:
            for layer in self.preprocess_layer:
                x = layer(x)
                if self.config["net"]["activation"] == "relu":
                    x = x.relu()
        i: int = 0
        for layer in self.conv_layers:
            # message
            x = layer(x, data.train_pos_edge_index, edge_attr = data.train_pos_edge_attr)
            # batch norm
            if self.config["net"]["normalization"] is not None:
                x = self.norms[i](x)
            # dropout
            x = F.dropout(x, p=self.config["net"]["dropout"], training=self.training)
            # activation
            x = self.activation(x)
            i = i + 1
        return x
```

```python
class AutoencoderNet(torch.nn.Module):
    '''
    overal autoencoder net as depicted in figure
    '''
    def __init__(self,  config: Dict[str, Union[float, int, str]]  = CONFIG_TEMPLATE_AUTOENCODER) -> None:
        super(AutoencoderNet, self).__init__()
        self.config: Dict[str, Union[float, int, str]]  = config
        # encoder and decoder
        self.encoder_real: Encoder = Encoder(config)
        self.encoder_fake: Encoder = Encoder(config)
        self.AE_real: GAE = GAE(self.encoder_real)
        self.AE_fake: GAE = GAE(self.encoder_fake)
        # post process layer
        self.postprocess_layer = None
        output = 2
        if self.config["net"]["n_postprocess_layer"] != 0:
            self.postprocess_layer: nn.ModuleList = nn.ModuleList()
            self.norms: nn.ModuleList = nn.ModuleList()
            input = 2*config["net"]["out_channels"]
            for i in range(self.config["net"]["n_postprocess_layer"]):
                self.postprocess_layer.append(nn.Linear(input, output if i == self.config["net"]["n_postprocess_layer"] - 1 else int(input/2)))
                input = int(input/2)
                self.norms.append(torch.nn.BatchNorm1d(input if i != self.config["net"]["n_postprocess_layer"] - 1 else output))
        # activation
        if self.config["net"]["activation"] == "relu":
            self.activation: torch.nn.ReLU = torch.nn.ReLU()
        elif self.config["net"]["activation"] == "prelu":
            self.activation: torch.nn.PReLU = torch.nn.PReLU(num_parameters = self.config["net"]["prelu_num_parameters"],
                                                             init = self.config["net"]["prelu_init"])
        elif self.config["net"]["activation"] == "swish":
          self.activation: torch.nn.SiLU = torch.nn.SiLU()
        self.dropout: torch.nn.Dropout = torch.nn.Dropout(p = 0.2)
        return None

    def forward(self, data_real: torch_geometric.data.Data, data_fake: torch_geometric.data.Data) -> List[torch_geometric.data.Data]:
        # retrieve encoder output
        recon_real: torch_geometric.data.Data = self.AE_real.encode(data_real)
        recon_fake: torch_geometric.data.Data = self.AE_fake.encode(data_fake)
        # retrieve encoder output
        output_real: torch_geometric.data.Data = recon_real.clone()
        output_fake: torch_geometric.data.Data = recon_fake.clone()
        # apply global mean
        if self.config["net"]["aggregation"] == "mean":
            output_real: torch.tensor  = global_mean_pool(output_real, data_real.batch)
            output_fake: torch.tensor  = global_mean_pool(output_fake, data_fake.batch)
        elif self.config["net"]["aggregation"] == "add":
            output_real: torch.tensor  = global_add_pool(output_real, data_real.batch)
            output_fake: torch.tensor  = global_add_pool(output_fake, data_fake.batch)
        elif self.config["net"]["aggregation"] == "max":
            output_real: torch.tensor  = global_max_pool(output_real, data_real.batch)
            output_fake: torch.tensor  = global_max_pool(output_fake, data_fake.batch)
        output: torch.tensor = torch.cat([output_real.clone().detach(), output_fake.clone().detach()], dim = 1)
        # Apply post processing layers
        i: int = 0
        for layer in self.postprocess_layer:
            output = layer(output)
            if i != self.config["net"]["n_postprocess_layer"] - 1:
                output = self.activation(output)
        return recon_real, recon_fake, output_real, output_fake, output


    def forward_autoencoder(self,  data_real: torch_geometric.data.Data,
                            data_fake: torch_geometric.data.Data) -> List[torch_geometric.data.Data]:
        '''
        forward only in autoencoder
        '''
        # retrieve encoder output
        recon_real: torch_geometric.data.Data = self.AE_real.encode(data_real)
        recon_fake: torch_geometric.data.Data = self.AE_fake.encode(data_fake)
        # retrieve encoder output
        output_real: torch_geometric.data.Data = recon_real
        output_fake: torch_geometric.data.Data = recon_fake
        # apply global mean
        if self.config["net"]["aggregation"] == "mean":
            output_real: torch.tensor = global_mean_pool(output_real, data_real.batch)
            output_fake: torch.tensor  = global_mean_pool(output_fake, data_fake.batch)
        elif self.config["net"]["aggregation"] == "add":
            output_real: torch.tensor  = global_add_pool(output_real, data_real.batch)
            output_fake: torch.tensor  = global_add_pool(output_fake, data_fake.batch)
        elif self.config["net"]["aggregation"] == "max":
            output_real: torch.tensor  = global_max_pool(output_real, data_real.batch)
            output_fake: torch.tensor  = global_max_pool(output_fake, data_fake.batch)
        output: torch.tensor  = torch.cat([output_real.clone().detach(), output_fake.clone().detach()], dim = 1)
        return recon_real, recon_fake, output_real, output_fake, output

    def forward_classifier(self, input: torch.tensor):
        '''
        forward only in classifier
        '''
        # Apply post processing layers
        i: int = 0
        for layer in self.postprocess_layer:
            input: torch.tensor = layer(input)
            input = self.norms[i](input)
            if i != self.config["net"]["n_postprocess_layer"] - 1:
                input = self.activation(input)
            input = self.dropout(input)
            i = i + 1
        output = input
        return  output
```

```python
class MessageBlockReinf(torch.nn.Module):
  '''
  message block similar to predictor net
  '''
  def __init__(self, num_node_features, num_edge_features, config: Dict[str, Union[int, str, float]]  = CONFIG_TEMPLATE_REINF["actor"], i = 0):
      super(MessageBlockReinf, self).__init__()
      self.config: Dict[str, Union[int, str, float]]  = config
      self.num_node_features: int = num_node_features
      self.num_edge_features: int = num_edge_features
      self.hidden_channels: int = config["net"]["hidden_channels"]
      # activation
      if self.config["net"]["activation"] == "relu":
          self.activation: torch.nn.ReLU = torch.nn.ReLU()
      elif self.config["net"]["activation"] == "prelu":
          self.activation: torch.nn.PReLU = torch.nn.PReLU(num_parameters=self.config["net"]["prelu_num_parameters"], init=self.config["net"]["prelu_init"])
      elif self.config["net"]["activation"] == "swish":
        self.activation: torch.nn.SiLU = torch.nn.SiLU()
      # convolution
      self.conv = None
      if self.config["net"]["type_conv"] == "gat":
          self.conv: GATConv = GATConv(self.num_node_features if i == 0 else self.hidden_channels, self.hidden_channels, edge_dim = self.num_edge_features,
                              negative_slope = self.config["net"]["negative_slope"]) # ignore dropout here given we apply it thereafter
      elif self.config["net"]["type_conv"] == "reset":
          self.conv: ResGatedGraphConv = ResGatedGraphConv(self.num_node_features if i == 0 else self.hidden_channels, self.hidden_channels, edge_dim = self.num_edge_features,
                              act = self.activation) # pass same activation
      elif self.config["net"]["type_conv"] == "conv":
          self.conv: GATv2Conv = GATv2Conv(self.num_node_features if i == 0 else self.hidden_channels, self.hidden_channels, edge_dim = self.num_edge_features,
                                negative_slope = self.config["net"]["negative_slope"]) # pass same activation
      elif self.config["net"]["type_conv"] == "pdn":
          self.conv: PDNConv = PDNConv(self.num_node_features if i == 0 else self.hidden_channels,
                              self.hidden_channels, edge_dim = self.num_edge_features,
                              hidden_channels = self.hidden_channels )
      # normalization
      if self.config["net"]["normalization"] is not None:
          if self.config["net"]["normalization"] == "batch":
              self.norm: BatchNorm = BatchNorm(config["net"]["hidden_channels"] )
          elif self.config["net"]["normalization"] == "graph":
              self.norm: GraphNorm = GraphNorm(config["net"]["hidden_channels"] )
          if self.config["net"]["normalization"] == "pair":
              self.norm: PairNorm = PairNorm(config["net"]["hidden_channels"] )
          if self.config["net"]["normalization"] == "mean":
              self.norm: MeanSubtractionNorm = MeanSubtractionNorm( )

  def forward(self, x: torch.tensor, data: torch_geometric.data.Data) -> List[torch.tensor]:
      # message
      x: torch.tensor = self.conv(x, data.edge_index, edge_attr = data.edge_attr)
      # batch norm
      if self.config["net"]["normalization"] is not None:
          x = self.norm(x)
      # dropout
      x = F.dropout(x, p=self.config["net"]["dropout"], training=self.training)
      # activation
      x = self.activation(x)
      return x
```

```python
class BranchNet(torch.nn.Module):
    '''
    handle a graph similarly to a predictor net
    '''
    def __init__(self, num_node_features, num_edge_features, config: Dict[str, Union[int, str, float]]  = CONFIG_TEMPLATE_REINF["actor"]) -> None:
        super(BranchNet, self).__init__()
        self.config: Dict[str, Union[int, str, float]]  = config
        self.num_node_features: int = num_node_features
        self.num_edge_features: int = num_edge_features
        self.hidden_channels: int = config["net"]["hidden_channels"]
        # preprocess layer
        self.preprocess_layer: nn.ModuleList = None
        if self.config["net"]["n_preprocess_layer"] != 0:
            self.preprocess_layer = nn.ModuleList([nn.Linear(self.num_node_features, self.num_node_features) for i in range(self.config["net"]["n_preprocess_layer"])])
        # message passing layer
        self.messag_passing_layers: nn.ModuleList = None
        if self.config["net"]["n_message_passing_layer"] != 0:
            self.messag_passing_layers = nn.ModuleList([MessageBlockReinf(self.num_node_features, self.num_edge_features, config, i) for i in range(self.config["net"]["n_message_passing_layer"])])
        # jumping knowledge
        if self.config["net"]["jumping_knowledge"] is not None:
            self.jk: JumpingKnowledge = JumpingKnowledge(
                self.config["net"]["jumping_knowledge"],
                self.hidden_channels, self.config["net"]["n_message_passing_layer"])
        # activation
        if self.config["net"]["activation"] == "relu":
            self.activation: torch.nn.ReLU = torch.nn.ReLU()
        elif self.config["net"]["activation"] == "prelu":
            self.activation: torch.nn.PReLU = torch.nn.PReLU(
                num_parameters = self.config["net"]["prelu_num_parameters"],
                init = self.config["net"]["prelu_init"])
        elif self.config["net"]["activation"] == "swish":
            self.activation: torch.nn.SiLU = torch.nn.SiLU()
        return None

    def forward(self, data: torch_geometric.data.Data) -> List[torch.tensor]:
        x: torch.tensor = data.x
        # Apply preprocess layer
        for layer in self.preprocess_layer:
            x = layer(x)
            x = self.activation(x)
        # Apply message passing blocks
        xs: List[torch.tensor] = []
        i: int = 0
        for layer in self.messag_passing_layers:
            pre_x = x.clone()
            x = layer(x, data)
            xs.append(x)
            i = i + 1
        # apply jumping knowledge
        if self.config["net"]["jumping_knowledge"] is not None:
           x = self.jk(xs)
        # apply global mean
        if self.config["net"]["aggregation"] == "mean":
            pooled_x = global_mean_pool(x, data.batch)
        elif self.config["net"]["aggregation"] == "add":
            pooled_x = global_add_pool(x, data.batch)
        elif self.config["net"]["aggregation"] == "max":
            pooled_x = global_max_pool(x, data.batch)
        return x, pooled_x
```

```python
class ActorNet(torch.nn.Module):
    '''
    actor network
    '''
    def __init__(self,  config: Dict[str, Union[int, str, float]]  = CONFIG_TEMPLATE_REINF["actor"]) -> None:
        super(ActorNet, self).__init__()
        self.config: Dict[str, Union[int, str, float]]  = config
        self.hidden_channels: int = config["net"]["hidden_channels"]
        self.computed_features: int = 30
        # create two branch net
        self.original_graph_branch: BranchNet = BranchNet(1, 1,  config = self.config)
        self.embed_branch: BranchNet = BranchNet(3, 3, config = self.config)
        # activation
        if self.config["net"]["activation"] == "relu":
            self.activation: torch.nn.ReLU = torch.nn.ReLU()
        elif self.config["net"]["activation"] == "prelu":
            self.activation: torch.nn.PReLU = torch.nn.PReLU(num_parameters=self.config["net"]["prelu_num_parameters"], init=self.config["net"]["prelu_init"])
        elif self.config["net"]["activation"] == "swish":
          self.activation: torch.nn.SiLU = torch.nn.SiLU()
        # post process layer
        input: int = 0
        if self.config["net"]["jumping_knowledge"] is None:
              input = 2*self.hidden_channels + self.computed_features
        else:
              input = 2*self.config["net"]["n_message_passing_layer"]*self.hidden_channels + self.computed_features
        self.action_type_linear: nn.Linear = nn.Linear(input, 3)  # linear layer for action type
        self.first_node_linear: nn.Linear =  nn.Linear(2*self.hidden_channels, 1) # linear layer for first node selected
        self.second_node_linear_add: nn.Linear =  nn.Linear(2*self.hidden_channels, 1)  # linear layer for second node selected in add action
        self.second_node_linear_remove: nn.Linear =  nn.Linear(3*self.hidden_channels, 1)  # linear layer for second node selected in subtraact action
        return None


    def forward(self, observation: Dict[str, Union[torch_geometric.data.Data, torch.tensor, List[int]]]) -> Dict[str, torch.tensor]:
        original_x, original_x_pooled  = self.original_graph_branch(observation["Input_Graph"])
        embed_x, embed_x_pooled = self.embed_branch(observation["Embed_Plus_Scaffold_Graph"])
        # action construction
        action: Dict[str, torch.tensor] = copy.deepcopy(ACTION_DICT)
        reshaped_features: torch.tensor = torch.reshape(observation["Features"], (1, self.computed_features))
        reshaped_features: torch.tensor = reshaped_features.type(torch.FloatTensor)
        agg_x: torch.tensor = torch.cat([original_x_pooled, embed_x_pooled, reshaped_features.to(DEVICE)], dim=1)
        action["action_type"] = self.action_type_linear(agg_x).squeeze(0)
        action["first_node"] = torch.tensor([])
        action["second_node"] = torch.tensor([])
        action_type = action["action_type"].argmax(dim=0).item()
        if action_type != 2: # not terminate
            # concat embed and bin features
            embed_first: List[torch.tensor] = []
            reverse_taken: List[int] = copy.deepcopy(observation["taken_index"])
            reverse_taken.sort(reverse = True)
            for index in reverse_taken:
                emb: torch.tensor = embed_x[index, :]
                original: int = original_x[observation["bin_associator"][index], :]
                embed_first.append(torch.cat([emb, original]))
            embed_first = torch.stack(embed_first)
            first_node: torch.tensor = self.first_node_linear(embed_first)
            first_node = first_node.squeeze(1)
            action["first_node"] = first_node
            first_node = first_node.argmax(dim=0).item()
            if action_type == 0: # add edge
                # concat embed and first selected features
                embed_second: List[torch.tensor] = []
                reverse_taken: List[int] = copy.deepcopy(observation["remain_index"])
                reverse_taken.sort(reverse = True)
                for index in reverse_taken:
                    emb: torch.tensor = embed_x[index, :]
                    selected: int = embed_x[first_node, :]
                    embed_second.append(torch.cat([emb, selected]))
                embed_second: torch.tensor = torch.stack(embed_second)
                second_node: torch.tensor = self.second_node_linear_add(embed_second)
                second_node: torch.tensor = second_node.squeeze(1)
                action["second_node"] = second_node
            elif action_type == 1: # remove edge
                # concat embed, bin and first selected features
                embed_second: List[torch.tensor] = []
                reverse_taken: List[int] = copy.deepcopy(observation["taken_index"])
                reverse_taken.sort(reverse = True)
                for index in reverse_taken:
                    emb: torch.tensor = embed_x[index, :]
                    original: torch.tensor = original_x[observation["bin_associator"][index], :]
                    selected: torch.tensor = embed_x[first_node, :]
                    embed_second.append(torch.cat([emb, original, selected]))
                embed_second: torch.tensor = torch.stack(embed_second)
                second_node: torch.tensor = self.second_node_linear_remove(embed_second)
                second_node: torch.tensor = second_node.squeeze(1)
                action["second_node"] = second_node
        return action
```

```python
class CriticNet(torch.nn.Module):
    '''
    critic network
    '''
    def __init__(self,  config: Dict[str, Union[int, str, float]]  = CONFIG_TEMPLATE_REINF["critic"])  -> None:
        super(CriticNet, self).__init__()
        self.config: Dict[str, Union[int, str, float]]  = config
        self.hidden_channels: int = config["net"]["hidden_channels"]
        self.computed_features: int = 30
        # create two branch net
        self.original_graph_branch: BranchNet = BranchNet(1, 1,  config = self.config)
        self.embed_branch: BranchNet = BranchNet(3, 3, config = self.config)
        # activation
        if self.config["net"]["activation"] == "relu":
            self.activation:  torch.nn.ReLU = torch.nn.ReLU()
        elif self.config["net"]["activation"] == "prelu":
            self.activation: torch.nn.PReLU = torch.nn.PReLU(
                num_parameters = self.config["net"]["prelu_num_parameters"],
                init = self.config["net"]["prelu_init"])
        elif self.config["net"]["activation"] == "swish":
          self.activation: torch.nn.SiLU = torch.nn.SiLU()
        # post process layer
        input: int = 0
        if self.config["net"]["jumping_knowledge"] is None:
              input = 2*self.hidden_channels + self.computed_features
        else:
              input = 2*self.config["net"]["n_message_passing_layer"]*self.hidden_channels + self.computed_features
        self.critic: nn.Linear  = nn.Linear(input, 1)  # linear layer for action type
        return None

    def forward(self, observation) -> torch.tensor:
        original_x, original_x_pooled  = self.original_graph_branch(observation["Input_Graph"])
        embed_x, embed_x_pooled = self.embed_branch(observation["Embed_Plus_Scaffold_Graph"])
        # action construction
        action: Dict[str, torch.tensor] = copy.deepcopy(ACTION_DICT)
        reshaped_features: torch.tensor = torch.reshape(observation["Features"], (1, self.computed_features))
        reshaped_features: torch.tensor = reshaped_features.type(torch.FloatTensor)
        agg_x: torch.tensor  = torch.cat([original_x_pooled, embed_x_pooled, reshaped_features.to(DEVICE)], dim=1)
        output: torch.tensor  = self.critic(agg_x)
        return output
```

```python
class DwaveEnvironment():
    '''
    class to handle reinforcement learning environment
    '''
    train_data: List[torch_geometric.data.Data] = REINFORCEMENT_DATA_TRAIN
    train_index: int = 0
    eval_data: List[torch_geometric.data.Data] = REINFORCEMENT_DATA_EVAL
    eval_index: int = 0
    test_data: List[torch_geometric.data.Data] = REINFORCEMENT_DATA_TEST
    test_index: int = 0
    adversarial_net: AutoencoderNet = AutoencoderNet().to(DEVICE) # autoencoder net for reward computation
    discriminator_net: PredictorNet = PredictorNet().to(DEVICE)  # discriminator net for reward computation
    discriminator_reward: int = 500
    invalid_reward: int = 100
    final_reward: int = 10000

    def __init__(self,  config: Dict[str, Union[float, int, str]] = CONFIG_TEMPLATE_REINF["environment"]) -> None :
        self.config: Dict[str, Union[float, int, str]] = config # take config template for further optimization
        # attributes mantained during episodes
        # set train and test index to zero at start
        DwaveEnvironment.train_index = 0
        DwaveEnvironment.test_index = 0
        # uniform torque compensation calculator
        self.strength_computer =  partial(
            uniform_torque_compensation, prefactor=random.uniform(MINIMUM_TORQUE_VALUE, MAXIMUM_TORQUE_VALUE)) # create alias to compensation
        # target structure searcher
        self.child_structure_search: networkx.Graph = dimod.child_structure_dfs
        # get available qpu
        self.qpu_advantage: DWaveSampler = DWaveSampler()
        # get the target structure of qpu
        target_structure = self.child_structure_search(self.qpu_advantage)
        self.target_nodelist, self.target_edgelist, self.target_adjacency = target_structure
        self.translator: Dict[int, int] = {}
        ordered_nodelist = copy.deepcopy(self.target_nodelist)
        ordered_nodelist.sort()
        self.translator["emb_index_to_linear_index"] = {}
        self.translator["linear_index_to_emb_index"] = {}
        for i in range(len(ordered_nodelist)):
            self.translator["emb_index_to_linear_index"][ordered_nodelist[i]] = i
            self.translator["linear_index_to_emb_index"][i] = ordered_nodelist[i]
        self.annealing_time: int =  random.randint(MINIMUM_ANNEAL_TIME, MAXIMUM_ANNEAL_TIME)
        # attributes to be mantained in a single episode
        self.data_item: Dict[str, Union[int, torch.tensor, torch_geometric.data.Data]] = None  # qubo problem to be solved
        self.Q: Dict[str, int] = None # Q matrix form of problem
        self.bqm = None # bqm problem formulation
        self.default_strength: float = None # strength of chain for problem
        self.starting_embed = None # first minorminer embed founded
        self.actual_embed = None # last embed founder
        # factors used to comput auto scale
        self.max_h_range: float = None
        self.min_h_range: float = None
        self.max_j_range: float = None
        self.min_j_range: float = None
        self.remaining_nodelist: List[int] = None # remaining nodes available
        self.original_data: networkx.Graph = None # this is an attribute use to contain first embed networkx data produced in an episode
        self.phase: str = "train"
        self.train_rewards()
        self.adversarial_optimizer = None
        self.discriminator_optimizer = None
        return None

    def use_pretrain(self) -> None:
        '''
        use pretrain reward netowrkx
        '''
        DwaveEnvironment.adversarial_net.load_state_dict(torch.load(PATH_ADVERSARIAL, map_location = DEVICE))
        DwaveEnvironment.discriminator_net.load_state_dict(torch.load(PATH_DISCRIMINATOR, map_location = DEVICE))
        return None

    def use_pretrain_reinf(self) -> None:
        '''
        use pretrained reward networks learned during ppo
        '''
        DwaveEnvironment.adversarial_net.load_state_dict(torch.load(PATH_ADVERSARIAL_REINF, map_location = DEVICE))
        DwaveEnvironment.discriminator_net.load_state_dict(torch.load(PATH_DISCRIMINATOR_REINF, map_location = DEVICE))
        return None

    def train_rewards(self) -> None:
        '''
        set reward network in train mode
        '''
        DwaveEnvironment.adversarial_net.train()
        DwaveEnvironment.discriminator_net.train()
        return None

    def test_rewards(self) -> None:
        '''
        set reward netowkrs in test mode
        '''
        DwaveEnvironment.adversarial_net.eval()
        DwaveEnvironment.discriminator_net.eval()
        return None

    def create_correct_op(self, num_ops: int = 10) -> torch_geometric.data.Data:
        '''
        create a graph with only valid operation preserving embedding correctness
        '''
        # set to none some element
        reference_graph: torch_geometric.data.Data = self.original_data
        reference_graph_nx: networkx.Graph = torch_geometric.utils.convert.to_networkx(
            reference_graph, to_undirected= True, edge_attrs = ["edge_attr"], node_attrs = ["x"]) # nx version of the reference graph
        max_strength: float = max(
            [torch.max(reference_graph.x[:, 0]).item(), torch.max(reference_graph.x[:, 0]).item()]) # maximum strength inside graph
        index_list: List[int] = reference_graph.x[:, 1].tolist()
        index_set: Set[int] = set( index_list) # index set
        edge_set: Set[Tuple[int]] = set(list(reference_graph_nx.edges()))
        index_dict: Dict[int, int] = {i:index_list[i] for i in range(len(index_list))}
        max_index: int = max(index_set) # maximum index inside graph
        # want always at least a valid variation
        # do random number of operation
        extracted_num_op: int = random.randrange(1, num_ops + 1)
        altered_graph: networkx.Graph = copy.deepcopy(reference_graph_nx)
        for k in range(extracted_num_op):
            choose: int = random.randrange(0, 2) # only choose third option if not all valid
            # add a node
            if choose == 0:
                new_index: int = random.randrange(max_index, 2*max_index)
                altered_graph.add_node(altered_graph.number_of_nodes(), x = [1.0, new_index])
                first: int = altered_graph.number_of_nodes() - 1
                second: int = random.choice(list(index_dict.keys()))
                while ((second == first) and ((first, second) not in edge_set)):
                    second = random.choice(list(index_dict.keys()))
                altered_graph.add_edge(first, second, edge_attr=[random.uniform(max_strength, 2*max_strength), new_index, index_dict[second]])
            # multiply node and/or edge values
            if choose == 1:
              multiplier: float = random.uniform(0, 2*max_strength)
              for couple in list(edge_set):
                  altered_graph.edges[couple]["edge_attr"][0] = multiplier*altered_graph.edges[couple]["edge_attr"][0]
              for i in range(len(index_list)):
                  altered_graph.nodes[i]["x"][0] = multiplier*altered_graph.nodes[i]["x"][0]
            index_list: List[int] = [altered_graph.nodes[i]["x"][1] for i in range(altered_graph.number_of_nodes())]
            index_set: Set[int] = set(index_list) # index set
            edge_set: Set[Tuple[int]] = set(list(altered_graph.edges()))
            index_dict: Dict[int, int] = {i:index_list[i] for i in range(len(index_list))}
            max_index: int = max(index_set) # maximum index inside graph
        return torch_geometric.utils.convert.from_networkx(altered_graph, ['x'], ['edge_attr'])

    def reset(self) -> Dict[str, torch_geometric.data.Data]:
        # INITIALIZE VALUE TO BE MANTAINED
        # take single data item
        if self.phase == "train":
            self.train_rewards()
            self.adversarial_optimizer = get_optimizer(DwaveEnvironment.adversarial_net, config = self.config)
            self.discriminator_optimizer = get_optimizer(DwaveEnvironment.discriminator_net, config = self.config)
            # take the data
            self.data_item = DwaveEnvironment.train_data[DwaveEnvironment.train_index]
            DwaveEnvironment.train_index = DwaveEnvironment.train_index + 1
            if DwaveEnvironment.train_index == len(DwaveEnvironment.train_data) - 1:
                DwaveEnvironment.train_index = 0
        elif self.phase == "eval":
            self.test_rewards()
            # take the data
            self.data_item = DwaveEnvironment.eval_data[DwaveEnvironment.eval_index]
            DwaveEnvironment.eval_index = DwaveEnvironment.eval_index + 1
            if DwaveEnvironment.eval_index == len(DwaveEnvironment.eval_data) - 1:
                DwaveEnvironment.eval_index = 0
        else:
            self.test_rewards()
            # take the data
            self.data_item = DwaveEnvironment.test_data[DwaveEnvironment.test_index]
            DwaveEnvironment.test_index = DwaveEnvironment.test_index + 1
            if DwaveEnvironment.test_index == len(DwaveEnvironment.test_data) - 1:
                DwaveEnvironment.test_index = 0
        # minor embed  oriringal graph
        starting_graph: networkx.Graph = torch_geometric.utils.convert.to_networkx(
            Data().from_dict(self.data_item["Input_Graph"]), node_attrs = ["x"], edge_attrs = ["edge_attr"], to_undirected = True)
        # get factors of qubo
        linear: Dict[str, int] = {}
        for node in list(starting_graph.nodes(data=True)):
            linear[(str(node[0]), str(node[0]))] = node[1]["x"][0]
        quadratic: Dict[str, int] = {}
        for edge in list(starting_graph.edges(data=True)):
            quadratic[(str(edge[0]), str(edge[1]))] = edge[2]["edge_attr"][0]
        self.Q: Dict[str, int] = {**linear, **quadratic}
        self.bqm: BinaryQuadraticModel = BinaryQuadraticModel.from_qubo(self.Q)  # get bqm
        self.default_strength: float = self.strength_computer(self.bqm)  # default chain strength applied
        source_edgelist: List[Tuple[int]] = list(self.bqm.quadratic) + [(v, v) for v in self.bqm.linear]  # add self-loops to edgelist to handle singleton variables
        self.starting_embed = minorminer.find_embedding(source_edgelist, self.target_edgelist)
        # retrieve maximum and minimum rangers for auto scalign computation
        properties = dict(child_properties=self.qpu_advantage.properties.copy())
        self.max_h_range: float = max(properties['child_properties']['h_range'])
        self.min_h_range: float = min(properties['child_properties']['h_range'])
        self.max_j_range: float = max(properties['child_properties']['j_range'])
        self.min_j_range: float = max(properties['child_properties']['j_range'])
        # OBSERVATION CONSTRUCTION
        observation: Dict[str, Union[torch.tensor, int, torch_geometric.data.Data]] = copy.deepcopy(OBSERVATION_DICT)
        self.actual_embed: Dict[str, int] = self.starting_embed   # actual embed considered sofar (without scaffolding)
        # retrieve all node available node on architecture at start
        self.remaining_nodelist: List[int] = copy.deepcopy(self.target_nodelist)
        observation["bin_associator"] = {}
        for label in self.starting_embed:
          for node in self.starting_embed[label]:
            self.remaining_nodelist.remove(node)
            observation["bin_associator"][self.translator["emb_index_to_linear_index"][node]] = int(label)
        # construct first observation
        input_graph: networkx.Graph = Data().from_dict(self.data_item["Input_Graph"]).to(DEVICE) # input graph to pytorch geometric format
        input_graph.x = input_graph.x.type(torch.FloatTensor).to(DEVICE)
        input_graph.edge_attr = input_graph.edge_attr.type(torch.FloatTensor).to(DEVICE)
        input_graph.ptr = torch.tensor([0, input_graph.x.shape[0]]).type(torch.LongTensor)
        self.input_graph_data = input_graph
        observation["Input_Graph"] = input_graph
        emb_structure: EmbeddedStructure = EmbeddedStructure(self.target_edgelist, self.starting_embed)
        structure = emb_structure.embed_bqm(self.bqm, chain_strength=self.default_strength,
                                           smear_vartype=dimod.SPIN)
        linear, quadratic, offset = structure.to_ising()  # transform to ising
        # find autoscale factor value
        factor1=max([value for value in linear.values()])/self.max_h_range if linear.values() else 0
        factor2=min([value for value in linear.values()])/self.min_h_range if linear.values() else 0
        factor3=max([value for value in quadratic.values()])/self.max_j_range if quadratic.values() else 0
        factor4=min([value for value in quadratic.values()])/self.min_j_range if quadratic.values() else 0
        autoscale_factor: float = max([max(factor1, 0), max(factor2, 0), max(factor3, 0), max(factor4, 0)])
        if autoscale_factor == 0:
            autoscale_factor: int = 1
        # convert to graph
        embed_graph: networkx.Graph = networkx.Graph()
        node_list: List[int] = []
        for label in linear:
            node_list.append((label, {"value": linear[label] / autoscale_factor, "mapping": label}))
        embed_graph.add_nodes_from(node_list)
        edge_list: List[Tuple[int]] = []
        for label in  quadratic:
            edge_list.append((label[0], label[1], {"value": quadratic[label] / autoscale_factor, "mapping_1": label[0], "mapping_2": label[1]}))
        embed_graph.add_edges_from(edge_list)
        feature_list: List[float] = []
        for lab in self.data_item['Input_Graph_Features']:
            feature_list.append(self.data_item['Input_Graph_Features'][lab] if self.data_item['Input_Graph_Features'][lab]  is not None else -1)
        # creating embed features
        feature_list.append(networkx.density(embed_graph))
        feature_list.append(embed_graph.number_of_nodes())
        feature_list.append(embed_graph.number_of_edges())
        spectrum=networkx.adjacency_spectrum(embed_graph)
        feature_list.append(abs(spectrum[0]) if len(spectrum)>0 else -1)
        feature_list.append(abs(spectrum[1]) if len(spectrum)>1 else -1)
        feature_list.append(abs(spectrum[2]) if len(spectrum)>2 else -1)
        feature_list.append(abs(spectrum[3]) if len(spectrum)>3 else -1)
        feature_list.append(abs(spectrum[4]) if len(spectrum)>4 else -1)
        largest: float = abs(spectrum[0])
        lowest: float = abs(spectrum[0])
        if (abs(spectrum[1]) if len(spectrum)>1 else -1) != -1:
            lowest = abs(spectrum[1])
        if (abs(spectrum[2]) if len(spectrum)>2 else -1) != -1:
            lowest = abs(spectrum[2])
        if (abs(spectrum[3]) if len(spectrum)>3 else -1) != -1:
            lowest = abs(spectrum[3])
        if (abs(spectrum[4]) if len(spectrum)>4 else -1) != -1:
            lowest = abs(spectrum[4])
        feature_list.append(largest-lowest)
        tr: int = 0
        triangles = networkx.triangles(embed_graph)
        for label in triangles:
          tr = tr + triangles[label]
        tr: int = tr/3 # each triangle is counted three times
        feature_list.append(tr)
        degree_list = [couple[1] for couple in embed_graph.degree]
        feature_list.append(min(degree_list))
        feature_list.append(max(degree_list))
        feature_list.append(sum(degree_list)/len(degree_list))
        feature_list.append(min([len(self.actual_embed) for label in self.actual_embed]))
        feature_list.append(max([len(self.actual_embed) for label in self.actual_embed]))
        feature_list.append(sum([len(self.actual_embed) for label in self.actual_embed]) / len([len(self.actual_embed) for label in self.actual_embed]))
        feature_list.append(self.default_strength)
        features: torch.tensor = torch.Tensor(feature_list).type(torch.FloatTensor)
        observation["Features"] = features.to(DEVICE)
        # construct embed graph plus scaffold
        # add bin to every node
        embed_graph_plus_scaffold: networkx.Graph = copy.deepcopy(embed_graph)
        full_edge_list: List[Tuple[int]] = []
        for edge in list(embed_graph_plus_scaffold.edges(data=True)):
            full_edge_list.append(
                (self.translator["emb_index_to_linear_index"][edge[0]], self.translator["emb_index_to_linear_index"][edge[1]], edge[2]))
        full_node_list: List[int] = []
        for node in list(embed_graph_plus_scaffold.nodes(data=True)):
            bin = None
            for label in self.actual_embed:
                for index in self.actual_embed[label]:
                    if index == node[0]:
                       bin = int(label)
            full_node_list.append(
                (self.translator["emb_index_to_linear_index"][node[0]],
                 {"value": node[1]["value"], "mapping": node[1]["mapping"], "bin": bin}))
        embed_graph_plus_scaffold: networkx.Graph  = networkx.Graph()
        # add non available node
        observation["remain_index"] = []
        for node in self.remaining_nodelist:
            full_node_list.append(
                (self.translator["emb_index_to_linear_index"][node], {"value": 0.0, "mapping": node, "bin": -1}))
            observation["remain_index"].append(self.translator["emb_index_to_linear_index"][node])
        observation["remain_index"].sort()
        observation["taken_index"] = []
        for label in self.actual_embed:
            for index in self.actual_embed[label]:
                observation["taken_index"].append(self.translator["emb_index_to_linear_index"][index])
        observation["taken_index"].sort()
        embed_graph_plus_scaffold.add_nodes_from(full_node_list)
        embed_graph_plus_scaffold.add_edges_from(full_edge_list)
        # transform to data item and add to observation
        if embed_graph_plus_scaffold.edges(data=True):
            embed_graph_plus_scaffold = torch_geometric.utils.convert.from_networkx(
                  embed_graph_plus_scaffold,
                  group_node_attrs  = ['value', 'mapping', "bin"],
                  group_edge_attrs = ['value', 'mapping_1', 'mapping_2']).to(DEVICE)
        else:
            embed_graph_plus_scaffold = torch_geometric.utils.convert.from_networkx(
                  embed_graph_plus_scaffold,
                  group_node_attrs  = ['value', 'mapping', "bin"]).to(DEVICE)
        if embed_graph_plus_scaffold.edge_attr is None:
            embed_graph_plus_scaffold.edge_attr = torch.full((embed_graph_plus_scaffold.edge_index.shape[1], 3), -1)
        embed_graph_plus_scaffold.x = embed_graph_plus_scaffold.x.type(torch.FloatTensor)
        embed_graph_plus_scaffold.edge_attr = embed_graph_plus_scaffold.edge_attr.type(torch.FloatTensor)
        embed_graph_plus_scaffold.ptr = torch.tensor([0, embed_graph_plus_scaffold.x.shape[0]]).type(torch.LongTensor)
        observation["Embed_Plus_Scaffold_Graph"] = embed_graph_plus_scaffold.to(DEVICE)
        # reward of discriminator
        if embed_graph.edges(data=True):
            data_discriminator = torch_geometric.utils.convert.from_networkx(
              embed_graph,
              group_node_attrs  = ['value', 'mapping'],
              group_edge_attrs = ['value', 'mapping_1', 'mapping_2']).to(DEVICE)
        else:
            data_discriminator = torch_geometric.utils.convert.from_networkx(
                embed_graph,
                group_node_attrs  = ['value', 'mapping']).to(DEVICE)
        if data_discriminator.edge_attr is None:
            data_discriminator.edge_attr = torch.full((data_discriminator.edge_index.shape[1], 3), -1)
        data_discriminator.features = observation["Features"]
        data_discriminator.x = data_discriminator.x.type(torch.FloatTensor)
        data_discriminator.edge_attr = data_discriminator.edge_attr.type(torch.FloatTensor)
        data_discriminator.features = data_discriminator.features.type(torch.DoubleTensor)
        data_discriminator.ptr = torch.tensor([0, data_discriminator.x.shape[0]]).type(torch.LongTensor)
        self.original_data = copy.deepcopy(data_discriminator).to(DEVICE) # this is used for autoencoder in all iterations
        self.previous_observation = observation
        self.previous_data_discriminator = data_discriminator
        return observation

    def anneal(self, embed) -> List[int]:
      '''
      anneal the given embed
      '''
      sampler: FixedEmbeddingComposite = FixedEmbeddingComposite(self.qpu_advantage, embedding=embed)
      sampleset: Dict[str, int] = sampler.sample_qubo(
          self.Q, num_reads=NUM_READS, return_embedding=True,
          auto_scale=True, chain_strength = self.default_strength,
          annealing_time=self.annealing_time)
      starting_graph: networkx.Graph = torch_geometric.utils.convert.to_networkx(
          Data().from_dict(self.data_item["Input_Graph"]),
          node_attrs = ["x"], edge_attrs = ["edge_attr"], to_undirected = True)
      complement: networkx.Graph = networkx.complement(starting_graph)
      actual_output_values: List[int] = []
      for label in sampleset.first[0]:
          if sampleset.first[0][label] == 1:
              actual_output_values.append(label)
      # output of regression and classification
      # note is necessary to check if it is a clique
      H: networkx.Graph = complement.subgraph(actual_output_values)
      n: int = len(actual_output_values)
      is_clique: bool = False
      regression_output: int = 0
      if H.size() == n*(n-1)/2:
          regression_output =  len(actual_output_values)
          is_clique = True
      else:
          regression_output = 0
      classification_output = 0
      if ((H.size() >= len(networkx.algorithms.approximation.clique.max_clique(complement))) and (is_clique == True)):
          classification_output = 1
      else:
          classification_output = 0
      return classification_output, regression_output

    def step(
        self,
        action: Dict[str, torch.tensor],
        force_termination: bool = False) -> List[Union[Dict[str, Union[torch.tensor, int, torch_geometric.data.Data]], int, float]]:
      '''
      do a step into env
      '''
      action_type: int = action["action_type"].argmax(dim=0)
      done: int = 0
      reward: float = 0
      observation: Dict[str, Union[torch.tensor, torch_geometric.data.Data, int]] = None
      if force_termination:
          action_type = 2
      if action_type == 2:
          # REWARD COMPUTATION
          # here call full evaluation
          classification_output, regression_output = self.anneal(self.actual_embed)
          discriminator_out: torch.tensor = DwaveEnvironment.discriminator_net(self.previous_data_discriminator.to(DEVICE))
          criterion: torch.nn.CrossEntropyLoss = torch.nn.CrossEntropyLoss()
          if self.phase == "train":
              loss: torch.nn.CrossEntropyLoss = criterion(
                  discriminator_out.to(DEVICE), torch.tensor([classification_output]).to(DEVICE)).to(DEVICE)
              loss.backward()
              self.discriminator_optimizer.step()
              self.discriminator_optimizer.zero_grad()
              reward = DwaveEnvironment.final_reward*(self.data_item['Input_Label']-regression_output)/self.data_item['Input_Label']
          done = self.data_item['Input_Label'] - regression_output
      else:
          rejected = False # check if action rejected
          if action_type == 0:  # add action
              first_node: int = action["first_node"].argmax(dim=0)
              second_node: int = action["second_node"].argmax(dim=0)
              taken_index: List[int]  = [self.translator["linear_index_to_emb_index"][self.previous_observation["taken_index"][i]] for i in range(len(self.previous_observation["taken_index"]))]
              taken_index.sort()
              remain_index: List[int] = [self.translator["linear_index_to_emb_index"][self.previous_observation["remain_index"][i]] for i in range(len(self.previous_observation["remain_index"]))]
              remain_index.sort()
              first_node: int = taken_index[first_node]
              second_node: int = remain_index[second_node]
              edge_set: Set[Tuple[int]] = set(self.target_edgelist)
              # - first and second node edge exist
              connection: Tuple[int] = [first_node, second_node]
              connection.sort()
              connection = tuple(connection)
              if connection not in edge_set:
                rejected = True
              if not(rejected):
                  # add second node to same bin first node
                  bin_first_node = None
                  for target in self.actual_embed:
                      for node in  self.actual_embed[target]:
                          if first_node == node:
                             bin_first_node = target
                  self.actual_embed[bin_first_node].append(second_node)
          elif action_type == 1: # remove action
              first_node: int = action["first_node"].argmax(dim=0)
              second_node: int = action["second_node"].argmax(dim=0)
              taken_index: List[int]  = [self.translator["linear_index_to_emb_index"][self.previous_observation["taken_index"][i]] for i in range(len(self.previous_observation["taken_index"]))]
              taken_index.sort()
              first_node: int = taken_index[first_node]
              second_node: int = taken_index[second_node]
              edge_set: Set[Tuple[int]] = set(self.target_edgelist)
              # - first and second node edge exist
              connection: Tuple[int] = [first_node, second_node]
              connection.sort()
              connection = tuple(connection)
              if connection not in edge_set:
                  rejected: bool = True
              else:
                  # - first and second node inside same target embedding
                  bin_first_node: int = None
                  bin_second_node: int = None
                  for target in self.actual_embed:
                      for node in  self.actual_embed[target]:
                          if first_node == node:
                             bin_first_node = target
                          if second_node == node:
                             bin_second_node = target
                  if bin_first_node != bin_second_node: # - two embedding that should have at least a connection should not become disconnected after that.
                      rejected = True
                  else:
                      proposed_embedding: Dict[str, int] = copy.deepcopy(self.actual_embed)
                      proposed_embedding[str(bin_first_node)].remove(second_node)
                      for couple in self.bqm.quadratic:
                          first_bin_node = proposed_embedding[couple[0]]
                          second_bin_node = proposed_embedding[couple[1]]
                          num_connection: int = 0
                          for node1 in first_bin_node:
                            for node2 in second_bin_node:
                                proposed_connection = [node1, node2]
                                proposed_connection.sort()
                                proposed_connection = tuple(proposed_connection)
                                if proposed_connection in edge_set:
                                   num_connection = num_connection + 1
                          if num_connection == 0:
                              rejected = True
              if not(rejected):
                  # remove second node from same bin first node
                  bin_first_node: int = None
                  for target in self.actual_embed:
                      for node in  self.actual_embed[str(target)]:
                          if first_node == node:
                              bin_first_node = target
                  self.actual_embed[str(bin_first_node)].remove(second_node)
          if not(rejected):
              # NEW STATE COMPUTATION
              # OBSERVATION CONSTRUCTION
              observation: Dict[str, Union[int, torch.tensor, torch_geometric.data.Data]] = copy.deepcopy(OBSERVATION_DICT)
              # retrieve all node available node on architecture at start
              self.remaining_nodelist: List[int] = copy.deepcopy(self.target_nodelist)
              observation["bin_associator"] = {}
              for label in self.actual_embed:
                  for node in self.actual_embed[label]:
                      self.remaining_nodelist.remove(node)
                      observation["bin_associator"][self.translator["emb_index_to_linear_index"][node]] = int(label)
              # construct first observation
              observation["Input_Graph"] = self.input_graph_data
              emb_structure: EmbeddedStructure = EmbeddedStructure(self.target_edgelist, self.actual_embed)
              structure = emb_structure.embed_bqm(self.bqm, chain_strength=self.default_strength,
                                                smear_vartype=dimod.SPIN)
              linear, quadratic, offset = structure.to_ising()  # transform to ising
              # find autoscale factor value
              factor1: float = max([value for value in linear.values()])/self.max_h_range if linear.values() else 0
              factor2: float = min([value for value in linear.values()])/self.min_h_range if linear.values() else 0
              factor3: float = max([value for value in quadratic.values()])/self.max_j_range if quadratic.values() else 0
              factor4: float = min([value for value in quadratic.values()])/self.min_j_range if quadratic.values() else 0
              autoscale_factor: float = max([max(factor1, 0), max(factor2, 0), max(factor3, 0), max(factor4, 0)])
              if autoscale_factor == 0:
                  autoscale_factor = 1
              # convert to graph
              embed_graph: networkx.Graph = networkx.Graph()
              node_list: List[int] = []
              for label in linear:
                  node_list.append((label, {"value": linear[label] / autoscale_factor, "mapping": label}))
              embed_graph.add_nodes_from(node_list)
              edge_list: List[Tuple[int]] = []
              for label in  quadratic:
                  edge_list.append(
                      (label[0], label[1], {"value": quadratic[label] / autoscale_factor, "mapping_1": label[0], "mapping_2": label[1]}))
              embed_graph.add_edges_from(edge_list)
              feature_list: float = []
              for lab in self.data_item['Input_Graph_Features']:
                  feature_list.append(
                      self.data_item['Input_Graph_Features'][lab] if self.data_item['Input_Graph_Features'][lab]  is not None else -1)
              # creating embed features
              feature_list.append(networkx.density(embed_graph))
              feature_list.append(embed_graph.number_of_nodes())
              feature_list.append(embed_graph.number_of_edges())
              spectrum: float = networkx.adjacency_spectrum(embed_graph)
              feature_list.append(abs(spectrum[0]) if len(spectrum)>0 else -1)
              feature_list.append(abs(spectrum[1]) if len(spectrum)>1 else -1)
              feature_list.append(abs(spectrum[2]) if len(spectrum)>2 else -1)
              feature_list.append(abs(spectrum[3]) if len(spectrum)>3 else -1)
              feature_list.append(abs(spectrum[4]) if len(spectrum)>4 else -1)
              largest: int = abs(spectrum[0])
              lowest: int = abs(spectrum[0])
              if (abs(spectrum[1]) if len(spectrum)>1 else -1) != -1:
                  lowest = abs(spectrum[1])
              if (abs(spectrum[2]) if len(spectrum)>2 else -1) != -1:
                  lowest = abs(spectrum[2])
              if (abs(spectrum[3]) if len(spectrum)>3 else -1) != -1:
                  lowest = abs(spectrum[3])
              if (abs(spectrum[4]) if len(spectrum)>4 else -1) != -1:
                  lowest = abs(spectrum[4])
              feature_list.append(largest-lowest)
              tr: int = 0
              triangles = networkx.triangles(embed_graph)
              for label in triangles:
                  tr = tr + triangles[label]
              tr = tr/3 # each triangle is counted three times
              feature_list.append(tr)
              degree_list = [couple[1] for couple in embed_graph.degree]
              feature_list.append(min(degree_list))
              feature_list.append(max(degree_list))
              feature_list.append(sum(degree_list)/len(degree_list))
              feature_list.append(min([len(self.actual_embed) for label in self.actual_embed]))
              feature_list.append(max([len(self.actual_embed) for label in self.actual_embed]))
              feature_list.append(
                  sum([len(self.actual_embed) for label in self.actual_embed]) / len([len(self.actual_embed) for label in self.actual_embed]))
              feature_list.append(self.default_strength)
              features: torch.tensor = torch.Tensor(feature_list).type(torch.FloatTensor)
              observation["Features"] = features.to(DEVICE)
              # construct embed graph plus scaffold
              # add bin to every node
              embed_graph_plus_scaffold: Dict[str, int] = copy.deepcopy(embed_graph)
              full_edge_list: List[Tuple[int]] = []
              for edge in list(embed_graph_plus_scaffold.edges(data=True)):
                  full_edge_list.append((self.translator["emb_index_to_linear_index"][edge[0]], self.translator["emb_index_to_linear_index"][edge[1]], edge[2]))
              full_node_list: List[int] = []
              for node in list(embed_graph_plus_scaffold.nodes(data=True)):
                  bin = None
                  for label in self.actual_embed:
                      for index in self.actual_embed[label]:
                          if index == node[0]:
                             bin = int(label)
                  full_node_list.append(
                      (self.translator["emb_index_to_linear_index"][node[0]],
                       {"value": node[1]["value"], "mapping": node[1]["mapping"], "bin": bin}))
              # add non available node
              observation["remain_index"] = []
              for node in self.remaining_nodelist:
                 full_node_list.append((self.translator["emb_index_to_linear_index"][node], {"value": 0.0, "mapping": node, "bin": -1}))
                 observation["remain_index"].append(self.translator["emb_index_to_linear_index"][node])
              observation["remain_index"].sort()
              observation["taken_index"] = []
              for label in self.actual_embed:
                  for index in self.actual_embed[label]:
                      observation["taken_index"].append(self.translator["emb_index_to_linear_index"][index])
              observation["taken_index"].sort()
              embed_graph_plus_scaffold: networkx.Graph  = networkx.Graph()
              embed_graph_plus_scaffold.add_nodes_from(full_node_list)
              embed_graph_plus_scaffold.add_edges_from(full_edge_list)
              # transform to data item and add to observation
              if embed_graph_plus_scaffold.edges(data=True):
                  embed_graph_plus_scaffold = torch_geometric.utils.convert.from_networkx(
                        embed_graph_plus_scaffold,
                        group_node_attrs  = ['value', 'mapping', "bin"],
                        group_edge_attrs = ['value', 'mapping_1', 'mapping_2']).to(DEVICE)
              else:
                  embed_graph_plus_scaffold = torch_geometric.utils.convert.from_networkx(
                        embed_graph_plus_scaffold,
                        group_node_attrs  = ['value', 'mapping', "bin"]).to(DEVICE)
              if embed_graph_plus_scaffold.edge_attr is None:
                  embed_graph_plus_scaffold.edge_attr = torch.full((embed_graph_plus_scaffold.edge_index.shape[1], 3), -1)
              embed_graph_plus_scaffold.x = embed_graph_plus_scaffold.x.type(torch.FloatTensor)
              embed_graph_plus_scaffold.edge_attr = embed_graph_plus_scaffold.edge_attr.type(torch.FloatTensor)
              embed_graph_plus_scaffold.ptr = torch.tensor([0, embed_graph_plus_scaffold.x.shape[0]]).type(torch.LongTensor)
              observation["Embed_Plus_Scaffold_Graph"] = embed_graph_plus_scaffold.to(DEVICE)
              # reward of discriminator
              if embed_graph.edges(data=True):
                  data_discriminator = torch_geometric.utils.convert.from_networkx(
                    embed_graph,
                    group_node_attrs  = ['value', 'mapping'],
                    group_edge_attrs = ['value', 'mapping_1', 'mapping_2']).to(DEVICE)
              else:
                  data_discriminator = torch_geometric.utils.convert.from_networkx(
                      embed_graph,
                      group_node_attrs  = ['value', 'mapping']).to(DEVICE)
              if data_discriminator.edge_attr is None:
                  data_discriminator.edge_attr = torch.full((data_discriminator.edge_index.shape[1], 3), -1)
              data_discriminator.features = observation["Features"]
              data_discriminator.x = data_discriminator.x.type(torch.FloatTensor)
              data_discriminator.edge_attr = data_discriminator.edge_attr.type(torch.FloatTensor)
              data_discriminator.features = data_discriminator.features.type(torch.DoubleTensor)
              data_discriminator.ptr = torch.tensor([0, data_discriminator.x.shape[0]]).type(torch.LongTensor)
              # REWARD CONSTRUCTION
              # reward of discriminator
              data_discriminator: torch_geometric.data.Data = torch_geometric.utils.convert.from_networkx(
                  embed_graph,
                  group_node_attrs  = ['value', 'mapping'],
                  group_edge_attrs = ['value', 'mapping_1', 'mapping_2']).to(DEVICE)
              data_discriminator.features = observation["Features"]
              data_discriminator.x = data_discriminator.x.type(torch.FloatTensor)
              data_discriminator.edge_attr = data_discriminator.edge_attr.type(torch.FloatTensor)
              data_discriminator.features = data_discriminator.features.type(torch.DoubleTensor)
              data_discriminator.ptr = torch.tensor([0, data_discriminator.x.shape[0]]).type(torch.LongTensor)
              discriminator_out = DwaveEnvironment.discriminator_net(data_discriminator.to(DEVICE)).argmax(dim=1).item()
              reward += DwaveEnvironment.discriminator_reward if discriminator_out == 1 else 0
              # autoencoder reward -   embed1 / grafo generato + embed1 / embed2  SEMPRE CHE PIU CHIAMATE ALL EMBED DIANO RISULTATI DIVERSI, SE NO USIAMO LA STESSA STRATEGIA IN PRECEDENZA
              # create a correct op graph
              correct_op: torch_geometric.data.Data = self.create_correct_op().to(DEVICE)
              correct_op.train_pos_edge_index = correct_op.edge_index
              correct_op.train_pos_edge_attr = correct_op.edge_attr
              original_data_autoencoder = copy.deepcopy(self.original_data).to(DEVICE)
              original_data_autoencoder.train_pos_edge_index = original_data_autoencoder.edge_index
              original_data_autoencoder.train_pos_edge_attr = original_data_autoencoder.edge_attr
              data_discriminator_outencoder = copy.deepcopy(data_discriminator).to(DEVICE)
              data_discriminator_outencoder.train_pos_edge_index = data_discriminator_outencoder.edge_index
              data_discriminator_outencoder.train_pos_edge_attr = data_discriminator_outencoder.edge_attr
              _, _,  _, _,  output_correct = DwaveEnvironment.adversarial_net(original_data_autoencoder, correct_op)
              _, _,  _, _,  output_uncorrect = DwaveEnvironment.adversarial_net(data_discriminator_outencoder, original_data_autoencoder)
              criterion: torch.nn.CrossEntropyLoss = torch.nn.CrossEntropyLoss()
              adversarial_reward = criterion(output_correct.to(DEVICE), torch.tensor([0]).to(DEVICE)).to(DEVICE) + criterion(output_uncorrect.to(DEVICE), torch.tensor([1]).to(DEVICE)).to(DEVICE)
              if self.phase == "train":
                  adversarial_reward.backward()
                  self.adversarial_optimizer.step()
                  self.adversarial_optimizer.zero_grad()
              reward = reward + adversarial_reward
              self.previous_observation = observation
              self.previous_data_discriminator = data_discriminator
          else: # if action rejected
             observation: Dict[str, Union[torch.tensor, int, torch_geometric.data.Data]] = self.previous_observation # return previous observation
             reward: float = -1*DwaveEnvironment.invalid_reward  # give negative reward for useless action
      return observation, reward, done
```

use different variant of adam optimizer. even if sgd can still achieve sota result, adam variant can achieve better result in less time comparably and are also used as standard in graph neural network training

```python
def get_optimizer(model: torch.nn.Module, lr: Optional[float] = None, config: Dict[str, Union[float, int]] = CONFIG_TEMPLATE_PREDICTOR):
    '''
    get optimizer
    '''
    if config["optimizer"]["optimizer_name"] == "AdamW":
        optimizer: torch.optim.AdamW =  torch.optim.AdamW(
            model.parameters(), lr = config["optimizer"]["lr"] if lr is None else lr,
            betas = config["optimizer"]["betas"], eps = config["optimizer"]["eps"],
            weight_decay = config["optimizer"]["weight_decay"])
    elif config["optimizer"]["optimizer_name"] == "Adamax":
        optimizer: torch.optim.Adamax =  torch.optim.Adamax(
            model.parameters(), lr = config["optimizer"]["lr"] if lr is None else lr,
            betas = config["optimizer"]["betas"], eps = config["optimizer"]["eps"],
            weight_decay = config["optimizer"]["weight_decay"])
    elif config["optimizer"]["optimizer_name"] == "NAdam":
        optimizer:  torch.optim.NAdam =  torch.optim.NAdam(
            model.parameters(), lr = config["optimizer"]["lr"] if lr is None else lr,
            betas = config["optimizer"]["betas"], eps = config["optimizer"]["eps"],
            weight_decay = config["optimizer"]["weight_decay"], momentum_decay = config["optimizer"]["momentum_decay"])
    elif config["optimizer"]["optimizer_name"] == "RAdam":
        optimizer: torch.optim.RAdam =  torch.optim.RAdam(
            model.parameters(), lr = config["optimizer"]["lr"] if lr is None else lr,
            betas = config["optimizer"]["betas"], eps = config["optimizer"]["eps"],
            weight_decay = config["optimizer"]["weight_decay"])
    return optimizer
```

#####**6.3.1 predictor**


```python
def train_predictor(model: torch.nn.Module, loader: DataLoader, criterion: torch.nn.CrossEntropyLoss,
                    optimizer: torch.optim.AdamW, epoch: int, task: str = "classification",
                    gradient_clipping: int = 1, scheduler = None) -> List[float]:
    '''
    train predictor
    '''
    model.train()
    loss_array: List[float] = []
    batch_idx: int = 1
    for batch in loader:
         batch: torch_geometric.data.Data = batch.to(DEVICE)
         out: torch.tensor = model(batch)
         label: torch.tensor = batch.y if task == "classification" else batch.y_regr.unsqueeze(1)
         loss: torch.tensor = criterion(out, label)
         loss.backward()
         loss_array.append(loss.clone().detach().cpu())
         if gradient_clipping is not None:
              torch.nn.utils.clip_grad_norm_(model.parameters(), gradient_clipping)
         optimizer.step()
         optimizer.zero_grad()
         if scheduler is not None:
                if type(scheduler) == torch.optim.lr_scheduler.CosineAnnealingWarmRestarts:
                    scheduler.step(epoch + batch_idx / batch_idx)
                elif type(scheduler) == torch.optim.lr_scheduler.CyclicLR:
                    scheduler.step()
                elif  type(scheduler) == torch.optim.lr_scheduler.ReduceLROnPlateau:
                    scheduler.step(loss)
         batch_idx: int = batch_idx + 1
    return loss_array
```

```python
def test_predictor(model: torch.nn.Module, loader: DataLoader, criterion = None, task: str = "classification") -> List[List[float]]:
     '''
     test predictor
     '''
     model.eval()
     cumulated_pred: List[int] = []
     cumulated_lab: List[int] =  []
     return_values: List[int] = []
     loss_array: List[float] = []
     criterion = criterion if task == "classification" else  torch.nn.MSELoss()
     for batch in loader:  # Iterate in batches over the training/test dataset.
         batch: torch_geometric.data.Data = batch.to(DEVICE)
         out: torch.tensor = model(batch)
         if ((criterion is not None) or (task == "regression")):
              label: int = batch.y if task == "classification" else batch.y_regr.unsqueeze(1)
              loss_array.append(criterion(out, label).clone().detach().cpu().item())
         cumulated_pred.extend(out.argmax(dim=1).tolist() if task == "classification" else out.tolist())
         cumulated_lab.extend(batch.y.tolist() if task == "classification" else batch.y_regr.tolist())
     if task == "classification":
          return_values.append(f1_score(cumulated_lab, cumulated_pred))
          return_values.append(accuracy_score(cumulated_lab, cumulated_pred))
     else:
          return_values.append(mean(loss_array))
     return return_values, loss_array
```

to evaluate binary classification performances crossentropy loss is used
cross entropy is known to be more confident than bcewithlogits  while outputting same results [28]
PoisonNLLLoss and GaussianNLLLoss suppose distribution data to have a specific shape.
label smoothing is also taken in consideration as sometimes improving performances


for regression smooth l1 loss is known to be less sensitive to outliers. allow to set bet


other losses are not used because not inherent to  task at hand
- multimargin loss for multiclass
- Hinge embedding loss and cosine loss is used for embedding
- Margin Ranking Loss is used for GAN or ranking task
- triplet margin loss is  employed more for information retrieval
- not mentioned losses are used more to match representation
or in case of Kullback-Leibler divergence, minimizing cross entropy is the same of minimizing kullback but with less computation




```python
def operative_loop_predictor(task: str = "classification", n_fold: int = 1,
                             n_epochs: int = 100, config: Dict[str, Union[float, int]] = CONFIG_TEMPLATE_PREDICTOR,
                             plot: bool = True) -> List[Union[List[float], torch.nn.Module]]:
    '''
    train and test a predictor model for multiple splits
    '''
    model: PredictorNet = PredictorNet(task = task, config = config).to(DEVICE)
    optimizer = get_optimizer(model, config = config)
    criterion: torch.nn.CrossEntropyLoss = torch.nn.CrossEntropyLoss(label_smoothing = config["loss"]["label_smoothing"]) if task == "classification" else torch.nn.SmoothL1Loss(beta=config["loss"]["beta"])
    train_loaders, dev_loaders, test_loaders   = get_loaders_predictor(task = task, batch_size = config["batch_size"],  n_fold = n_fold)
    test_metric_list: List[float] = []
    for i in range(len(train_loaders)):
        best_metric = 0 if task == "classification" else sys.maxsize
        best_model = None
        train_loader: DataLoader = train_loaders[i]
        dev_loader: DataLoader = dev_loaders[i]
        test_loader: DataLoader = test_loaders[i]
        pbar = tqdm(range(1,n_epochs))
        sampled_epochs: List[int] = []
        losses_train: List[float] = []
        losses_dev: List[float] = []
        # scheduler
        scheduler = None
        if config["scheduler"] != None:
            if config["scheduler"] == "cosine":
                scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(
                    optimizer, T_0 = config["scheduler_params"]["T_0"],
                    T_mult = config["scheduler_params"]["T_mult"], eta_min = config["scheduler_params"]["eta_min"])
            elif config["scheduler"] == "cyclic":
                scheduler = torch.optim.lr_scheduler.CyclicLR(
                    optimizer, base_lr = MIN_LR[task],
                    max_lr = MAX_LR[task],
                    step_size_up = config["scheduler_params"]["step_up"],
                    step_size_down = config["scheduler_params"]["step_down"],
                    gamma = config["scheduler_params"]["gamma"],
                    cycle_momentum = False if optimizer != "SGD" else True)
            elif config["scheduler"] == "plateau":
                scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
                    optimizer, "min", factor=config["scheduler_params"]["factor"],
                    patience=config["scheduler_params"]["patience"])
        for x in pbar:
            sampled_epochs.append(x)
            loss_train = train_predictor(model, train_loader, criterion, optimizer, x, task = task, scheduler = scheduler, gradient_clipping = config["gradient_clipping"])
            losses_train.append(np.asarray(loss_train).mean())
            dev_metric, loss_dev = test_predictor(model, dev_loader, criterion = criterion if task == "classification" else None, task = task)
            losses_dev.append(np.asarray(loss_dev).mean())
            if task == "classification":
                pbar.set_description(f'Epoch: {x:03d}, F1: {dev_metric[0]:.4f}, Accuracy: {dev_metric[1]:.4f}')
            else:
                pbar.set_description(f'Test metric partition: {i:03d}, Epoch: {x:03d}, MSE: {dev_metric[0]:.4f}')
            if (((task == "classification") and (dev_metric[0] > best_metric)) or ((task == "regression") and (dev_metric[0] < best_metric))):
              best_metric = best_metric
              best_model = model
        if best_model is None:
            best_model = model
        test_metric, _ = test_predictor(best_model, test_loader, task = task)
        print("Test metric partition " + str(i) + " is " + str(test_metric))
        test_metric_list.append(test_metric[0])
        if plot:
            plt.figure(num = 3, figsize=(8, 5)).patch.set_facecolor('white')
            plt.title('Train and Dev Losses')
            plt.ylabel('Loss')
            plt.xlabel('Epochs')
            plt.plot(sampled_epochs, losses_train, label='Train loss')
            plt.plot(sampled_epochs, losses_dev, label='Dev loss')
            plt.legend()
            plt.show()
        if len(train_loaders) == 1:
            return best_model, test_metric
    test_metric_array = np.asarray(test_metric_list)
    print('Result', round(test_metric_array.mean(), 3), '+-', round(test_metric_array.std(), 3))
    return test_metric_array, best_model
```

```python
def lr_range_test_predictor(
    task = "classification",
    n_epochs: int = 100,
    config: Dict[str, Union[float, int]] = CONFIG_TEMPLATE_PREDICTOR,
    plot: bool = True,
    lr: float = 0.0000001,
    mult: float = 1.5
    ) -> None:
    '''
    lr range test for predictor
    '''
    model: PredictorNet = PredictorNet(task = task, config = config).to(DEVICE)
    optimizer = get_optimizer(model, lr = lr, config = config)
    criterion: torch.nn.CrossEntropyLoss = torch.nn.CrossEntropyLoss(label_smoothing = config["loss"]["label_smoothing"]) if task == "classification" else torch.nn.SmoothL1Loss(beta=config["loss"]["beta"])
    # retrieve data
    train_loader, _, __   = get_loaders_predictor(task = task, n_fold = 1)
    # Print learning rate annotation
    print("********************************")
    print(f"LEARNING RATE FROM {lr}")
    print("--------------------------------")
    model.train()  # set model to train model
    loss_list: List[float] = []
    batch_idx: int = 0
    #scheduler
    scheduler: StepLR = StepLR(optimizer, step_size = 1, gamma = mult)
    i = 0
    for batch in train_loader:
        optimizer = get_optimizer(model, lr = lr, config = config)
        print("\n")
        print("--------------------------------")
        batch_idx += 1
        print("ITERATION " + str(batch_idx))
        optimizer.zero_grad(set_to_none = True)
        batch: torch_geometric.data.Data = batch.to(DEVICE)
        out: torch.tensor = model(batch)
        loss: torch.tensor = criterion(out, batch.y if task == "classification" else batch.y_regr)
        loss.backward()
        loss_list.append(loss.clone().detach().cpu())
        optimizer.step() # Update the weights
        scheduler.step()
        lr = lr * mult # update multiplier
        print("LR " + str(lr))
        print("LOSS " + str(loss))
        print("--------------------------------")
        i = i + 1
        if i == 100:
          break;
    # plotting
    if plot:
        plt.figure(num = 3, figsize=(8, 5)).patch.set_facecolor('white')
        plt.title('Train and Dev Losses')
        plt.ylabel('Loss')
        plt.xlabel('Step')
        steps = [i for i in range(len(loss_list))]
        plt.plot(steps, loss_list, label='Loss')
        plt.legend()
        plt.show()
    return None
```

```python
def train_autoencoder(model: torch.nn.Module, train_loader_real: List[torch_geometric.data.Data],
                      train_loader_fake:List[torch_geometric.data.Data], criterion: torch.nn.CrossEntropyLoss,
                      optimizer, labels : List[int],  epoch: int, scheduler = None,
                      batch_size: int = 32, gradient_clipping: int = None, head_wait: int = 30) -> np.ndarray:
    '''
    train autoencoder
    '''
    model.train()
    loss_array: List[float] = []
    i: int = 1
    labels_extr: List[int] = []
    outputs: List[int] = []
    wait: bool = True
    if epoch == head_wait:
      wait: bool = False
    for data_real in train_loader_real:
         data_real: torch_geometric.data.Data = data_real.to(DEVICE)
         data_fake: torch_geometric.data.Data = train_loader_fake[i-1].to(DEVICE)
         if batch_size != 1:
            recon_real, recon_fake,  output_real, output_fake,  output = model.forward_autoencoder(data_real, data_fake)
            label: torch.tensor = torch.tensor(labels[i-1])
            labels_extr.append(label)
            outputs.append(output)
            cosine: torch.nn.CosineEmbeddingLoss = torch.nn.CosineEmbeddingLoss()
            label_cosine: torch.tensor = torch.tensor(1).unsqueeze(0).to(DEVICE) if labels[i-1] == 0 else torch.tensor(-1).unsqueeze(0).to(DEVICE)
            loss: torch.tensor =  0.8*cosine(output_real, output_fake, label_cosine) + 0.2*(model.AE_real.recon_loss(recon_real, data_real.train_pos_edge_index) + model.AE_fake.recon_loss(recon_fake, data_fake.train_pos_edge_index))
            if gradient_clipping is not None:
                  torch.nn.utils.clip_grad_norm_(model.parameters(), gradient_clipping)
            if (((i % batch_size) == 0) and not(wait)):
                labels_extr = torch.stack(labels_extr).to(DEVICE)
                outputs: torch.tensor = torch.stack(outputs)
                outputs = outputs.squeeze(1)
                label_out = model.forward_classifier(outputs)
                loss: torch.tensor = loss +  criterion(label_out, labels_extr)
                labels_extr = []
                outputs = []
         else:
            if not(wait):
                cosine = torch.nn.CosineEmbeddingLoss()
                recon_real, recon_fake, output_real, output_fake, output = model(data_real, data_fake)
                label = torch.tensor(labels[i-1]).unsqueeze(0).to(DEVICE)
                label_cosine: torch.tensor = torch.tensor(1).unsqueeze(0).to(DEVICE) if labels[i-1] == 0 else torch.tensor(-1).unsqueeze(0).to(DEVICE)
                loss: torch.tensor = (
                    criterion(output, label)
                    + model.AE_real.recon_loss(recon_real, data_real.train_pos_edge_index)
                    + model.AE_fake.recon_loss(recon_fake, data_fake.train_pos_edge_index)
                    + cosine(output_real, output_fake, label_cosine)
                )
            else:
                cosine: torch.nn.CosineEmbeddingLoss = torch.nn.CosineEmbeddingLoss()
                recon_real, recon_fake, output_real, output_fake, output = model(data_real, data_fake)
                label_cosine: torch.tensor = torch.tensor(1).unsqueeze(0).to(DEVICE) if labels[i-1] == 0 else torch.tensor(-1).unsqueeze(0).to(DEVICE)
                loss: torch.tensor = (
                    model.AE_real.recon_loss(recon_real, data_real.train_pos_edge_index)
                    + model.AE_fake.recon_loss(recon_fake, data_fake.train_pos_edge_index)
                    + cosine(output_real, output_fake, label_cosine)
                )
         loss_array.append(loss.clone().detach().cpu())
         loss.backward(retain_graph=True)
         optimizer.step()
         optimizer.zero_grad()
         if scheduler is not None:
                if type(scheduler) == torch.optim.lr_scheduler.CosineAnnealingWarmRestarts:
                    scheduler.step(epoch + i / i)
                elif type(scheduler) == torch.optim.lr_scheduler.CyclicLR:
                    scheduler.step()
                elif  type(scheduler) == torch.optim.lr_scheduler.ReduceLROnPlateau:
                    scheduler.step(loss)
         i = i + 1
    return loss_array
```

```python
def test_autoencoder(model:torch.nn.Module, loader_real: List[torch_geometric.data.Data],
                     loader_fake: List[torch_geometric.data.Data], labels: List[int], criterion = None):
     '''
     test autoencoder
     '''
     model.eval()
     cumulated_pred: List[float] = []
     return_values: List[float] = []
     loss_array: List[float] = []
     auc_real_array: List[float] = []
     ap_real_array: List[float] = []
     auc_fake_array: List[float] = []
     ap_fake_array: List[float] = []
     i: int = 0
     for data_real in loader_real:  # Iterate in batches over the training/test dataset.
         data_real = data_real.to(DEVICE)
         data_fake = loader_fake[i].to(DEVICE)
         recon_real, recon_fake,  output_real, output_fake,  output = model(data_real, data_fake)
         label = torch.tensor(labels[i]).unsqueeze(0).to(DEVICE)
         label_cosine = torch.tensor(1).unsqueeze(0).to(DEVICE) if labels[i-1] == 0 else torch.tensor(-1).unsqueeze(0).to(DEVICE)
         if criterion is not None:
              cosine = torch.nn.CosineEmbeddingLoss()
              loss = criterion(output, label) + model.AE_real.recon_loss(recon_real, data_real.train_pos_edge_index) + cosine(output_real, output_fake, label_cosine) + model.AE_fake.recon_loss(recon_fake, data_fake.train_pos_edge_index)
              loss_array.append(loss.clone().detach().cpu())
         cumulated_pred.extend(output.argmax(dim=1).tolist())
         auc_real, ap_real = model.AE_real.test(recon_real, data_real.test_pos_edge_index , data_real.test_neg_edge_index)
         auc_real_array.append(auc_real)
         ap_real_array.append(ap_real)
         auc_fake, ap_fake = model.AE_fake.test(recon_fake, data_fake.test_pos_edge_index, data_fake.test_neg_edge_index)
         auc_fake_array.append(auc_fake)
         ap_fake_array.append(ap_fake)
         i = i + 1
     return_values.append(f1_score(labels, cumulated_pred))
     return_values.append(accuracy_score(labels, cumulated_pred))
     return_values.append(mean(auc_real_array))
     return_values.append(mean(ap_real_array))
     return_values.append(mean(auc_fake_array))
     return_values.append(mean(ap_fake_array))
     if criterion is None:
        return return_values
     else:
        return return_values, loss_array
```

```python
def operative_loop_autoencoder(n_epochs: int = 100, config: Dict[str, Union[float, int, str]] = CONFIG_TEMPLATE_AUTOENCODER, plot: bool = True) -> List[Union[List[int], torch.nn.Module]] :
    '''
    train and test autoencoder model
    '''
    model: AutoencoderNet = AutoencoderNet(config = config).to(DEVICE)
    optimizer = get_optimizer(model, config = config)
    criterion: torch.nn.CrossEntropyLoss = torch.nn.CrossEntropyLoss(label_smoothing = config["loss"]["label_smoothing"])
    train_loader_real, dev_loader_real, test_loader_real, train_loader_fake, dev_loader_fake, test_loader_fake, train_label, dev_label, test_label  = get_loaders_autoencoder()
    best_metric: int = 0
    best_model: torch.nn.module = None
    # scheduler
    batch_size = config["batch_size"]
    scheduler = None
    if config["scheduler"] != None:
        if config["scheduler"] == "cosine":
            scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(
                optimizer, T_0 = config["scheduler_params"]["T_0"],
                T_mult = config["scheduler_params"]["T_mult"], eta_min = config["scheduler_params"]["eta_min"])
        elif config["scheduler"] == "cyclic":
            scheduler = torch.optim.lr_scheduler.CyclicLR(
                optimizer, base_lr = MIN_LR["autoencoder"],
                max_lr = MAX_LR["autoencoder"],
                step_size_up = config["scheduler_params"]["step_up"],
                step_size_down = config["scheduler_params"]["step_down"],
                gamma = config["scheduler_params"]["gamma"],
                cycle_momentum = False if optimizer != "SGD" else True)
        elif config["scheduler"] == "plateau":
            scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
                optimizer, "min", factor=config["scheduler_params"]["factor"],
                patience=config["scheduler_params"]["patience"])
    pbar = tqdm(range(1,n_epochs))
    sampled_epochs: List[float] = []
    losses_train: List[float] = []
    losses_dev: List[float] = []
    for x in pbar:
        sampled_epochs.append(x)
        loss_train: torch.tensor = train_autoencoder(model, train_loader_real, train_loader_fake, criterion, optimizer, train_label, x, scheduler = scheduler, batch_size = batch_size, gradient_clipping = config["gradient_clipping"], head_wait = config["head_wait"])
        losses_train.append(np.asarray(loss_train).mean())
        dev_metric, loss_dev = test_autoencoder(model, dev_loader_real, dev_loader_fake, dev_label, criterion = criterion)
        losses_dev.append(np.asarray(loss_dev).mean())
        pbar.set_description(f'Epoch: {x:03d}, F1: {dev_metric[0]:.4f}, Accuracy: {dev_metric[1]:.4f}, AUC real: {dev_metric[2]:.4f}, AP real: {dev_metric[3]:.4f}, AUC fake: {dev_metric[4]:.4f}, AP fake: {dev_metric[5]:.4f}')
        print("HERE")
        if (dev_metric[0] > best_metric):
          best_metric: float = best_metric
          best_model: torch.nn.Module = copy.deepcopy(model)
    if best_model is None:
        best_model = model
    test_metric = test_autoencoder(best_model, test_loader_real,  test_loader_fake, test_label)
    if plot:
        plt.figure(num = 3, figsize=(8, 5)).patch.set_facecolor('white')
        plt.title('Train and Dev Losses')
        plt.ylabel('Loss')
        plt.xlabel('Epochs')
        plt.plot(sampled_epochs, losses_train, label='Train loss')
        plt.plot(sampled_epochs, losses_dev, label='Dev loss')
        plt.legend()
        plt.show()
    return test_metric, best_model
```

```python
PATH_ACTOR: str = "reinf_actor.pk"
PATH_CRITIC: str = "reinf_critic.pk"
PATH_EVAL_PERF: str = "eval_perf.pk"
PATH_ACTOR_LOSS: str = "actor_loss.pk"
PATH_CRITIC_LOSS: str = "critic_loss.pk"
```

```python
class PPO:
    '''
    handle ppo learning
    '''
    def __init__(self, config: Dict[str, Union[int, str, float]] = copy.deepcopy(CONFIG_TEMPLATE_REINF), use_saved_actor: bool = False, use_saved_env: bool = False) -> None:
        self.config: Dict[str, Union[int, str, float]]  = config  # save configuration
        self.env:  DwaveEnvironment =  DwaveEnvironment()  # create environment in train phase
        if use_saved_env:
            self.env.use_pretrain_reinf()
        # Initialize actor and critic networks
        if use_saved_actor:
            self.actor: ActorNet = ActorNet().to(DEVICE)
            self.actor.load_state_dict(torch.load(PATH_ACTOR, map_location = DEVICE))
            self.critic: CriticNet = CriticNet().to(DEVICE)
            self.critic.load_state_dict(torch.load(PATH_CRITIC, map_location = DEVICE))
        else:
            self.actor: ActorNet = ActorNet(config = config["actor"]).to(DEVICE)
            self.critic: CriticNet = CriticNet(config = config["critic"]).to(DEVICE)
        # optimizers
        self.actor_optim = get_optimizer(self.actor,  config = self.config["PPO"])
        self.critic_optim = get_optimizer(self.critic,  config = self.config["PPO"])
        # collector
        if use_saved_actor:
            self.eval_perf: List[float] = []
            self.actor_losses: List[float] = []
            self.critic_losses: List[float] = []
            with open(PATH_EVAL_PERF, "rb") as fp:   # Unpickling
                 self.eval_perf = pickle.load(fp)
            with open(PATH_ACTOR_LOSS, "rb") as fp:   # Unpickling
                 self.actor_losses = pickle.load(fp)
            with open(PATH_CRITIC_LOSS, "rb") as fp:   # Unpickling
                 self.critic_losses = pickle.load(fp)
        else:
            self.eval_perf: List[float] = []
            self.actor_losses: List[float] = []
            self.critic_losses: List[float] = []
        self.train_step: int = 0
        self.best_eval_sofar: float = 0
        return None

    def train_and_eval(self) -> None:
        '''
        do a train step and evaluate performance
        '''
        actor_loss_avg, critic_loss_avg = self.train()
        print("Actor loss avg: " + str(actor_loss_avg))
        print("Critic loss avg: " + str(critic_loss_avg))
        avg_perf, stdev = self.evaluate_perf()
        print("Avg perf eval: " + str(avg_perf))
        print("Stdev perf eval" + str(stdev))
        self.eval_perf.append(avg_perf)
        with open(PATH_EVAL_PERF, "wb") as fp:   #Pickling
           pickle.dump(self.eval_perf, fp)
        self.actor_losses.append(actor_loss_avg)
        with open(PATH_ACTOR_LOSS, "wb") as fp:   #Pickling
           pickle.dump(self.actor_losses, fp)
        self.critic_losses.append(critic_loss_avg)
        with open(PATH_CRITIC_LOSS, "wb") as fp:   #Pickling
           pickle.dump(self.critic_losses, fp)
        self.train_step = self.train_step + 1
        if avg_perf >= self.best_eval_sofar:
            torch.save(self.actor.state_dict(), PATH_ACTOR)
            torch.save(self.critic.state_dict(), PATH_CRITIC)
            torch.save(DwaveEnvironment.adversarial_net.state_dict(), PATH_ADVERSARIAL_REINF)
            torch.save(DwaveEnvironment.discriminator_net.state_dict(), PATH_DISCRIMINATOR_REINF)
            self.best_eval_sofar = avg_perf
        return None

    def plot(self, series: List[float], title: str = "actor performance") -> None:
        '''
        plot a performance graph
        '''
        plt.figure(num = 3, figsize=(8, 5)).patch.set_facecolor('white')
        plt.title(title)
        plt.ylabel('value')
        plt.xlabel('train step')
        plt.plot([i for i in range(len(series))], series)
        plt.legend()
        plt.show()
        return None

    def train(self) -> List[float]:
        '''
        single train step with ppo
        '''
        # set to train
        self.env.phase = "train"
        batch_observations, batch_actions, batch_log_probs, batch_rewards_to_go, batch_lens = self.collect_trajectories()  # collect trajectories (point 3)
        batch_log_probs: torch.tensor = torch.stack(batch_log_probs)
        value_function, _ = self.evaluate(batch_observations, batch_actions)  # compute V_{phi, k}
        advantage: torch.tensor = batch_rewards_to_go - value_function.detach()   # compute advantage estimates based on value function
        advantage = (advantage - advantage.mean()) / (advantage.std() + 1e-10)
        actor_losses: List[float] = []
        critic_losses: List[float] = []
        self.actor.train()
        self.critic.train()
        for _ in range(self.config["PPO"]["n_updates_per_iteration"]):
            value_function, curr_log_probs = self.evaluate(batch_observations, batch_actions) # compute pi_theta(a_t | s_t)
            curr_log_probs: torch.tensor = torch.stack(curr_log_probs)
            prob_ratio: torch.tensor = torch.exp(0.001*(curr_log_probs - batch_log_probs)) # compute probability ratio
            # actor update
            # surrogate losses
            surrogate_loss1: torch.tensor = prob_ratio * advantage
            surrogate_loss2: torch.tensor = torch.clamp(prob_ratio, 1 - self.config["PPO"]["clip"], 1 + self.config["PPO"]["clip"]) * advantage
            actor_loss: torch.tensor = (-torch.min(surrogate_loss1, surrogate_loss2)).mean()
            self.actor_optim.zero_grad()
            actor_loss.backward(retain_graph = True)
            self.actor_optim.step()
            # critic update
            mse: nn.MSELoss = nn.MSELoss()
            critic_loss: torch.tensor = mse(value_function, batch_rewards_to_go)
            self.critic_optim.zero_grad()
            critic_loss.backward()
            self.critic_optim.step()
            actor_losses.append(actor_loss.item())
            critic_losses.append(critic_loss.item())
        actor_loss_avg: float = sum(actor_losses)/len(actor_losses)
        critic_loss_avg: float = sum(critic_losses)/len(critic_losses)
        return actor_loss_avg, critic_loss_avg

    def evaluate(self, batch_observations: List[Dict[str, Union[torch.tensor, int, torch_geometric.data.Data]]], batch_actions: List[Dict[str, torch.tensor]]) -> List[torch.tensor]:
        '''
        compute value function
        '''
        value_function: List[torch.tensor] = []
        log_probs: List[torch.tensor] = []
        i: int = 0
        for observation in batch_observations:
            # Query critic network for a value V for each obs in batch_obs.
            out: torch.tensor = self.critic(observation)
            value_function.append(out)
            # Calculate the log probabilities of batch actions using most
            # recent actor network.
            # Query the actor network for a mean action.
            actor_action: Dict[str, torch.tensor] = self.actor(observation)  # query the actor network for a mean action.
            # extract log probs
            extended_first_node: torch.tensor =  torch.zeros(len(observation["taken_index"])).to(DEVICE)
            extended_first_node[:actor_action["first_node"].shape[0]] = actor_action["first_node"]
            extended_second_node = torch.zeros(len(observation["remain_index"])).to(DEVICE)
            extended_second_node[:actor_action["second_node"].shape[0]] = actor_action["second_node"]
            mean_tensor: torch.tensor = torch.cat([extended_first_node, extended_second_node, actor_action["action_type"]])
            extended_first_node: torch.tensor =  torch.zeros(len(observation["taken_index"])).to(DEVICE)
            extended_first_node[:batch_actions[i]["first_node"].shape[0]] = batch_actions[i]["first_node"]
            extended_second_node: torch.tensor = torch.zeros(len(observation["remain_index"])).to(DEVICE)
            extended_second_node[:batch_actions[i]["second_node"].shape[0]] = batch_actions[i]["second_node"]
            old_action_tensor: torch.tensor = torch.cat([extended_first_node, extended_second_node, batch_actions[i]["action_type"]])
            dist_action: MultivariateNormal  = MultivariateNormal(mean_tensor, torch.diag(torch.full(size=(mean_tensor.shape[0],), fill_value=1e-8)).to(DEVICE))
            log_probs.append(dist_action.log_prob(old_action_tensor.to(DEVICE)))
            i = i + 1
        value_function = torch.stack(value_function).squeeze().to(DEVICE) # tensor of shape [1, 2, 3]
        return value_function, log_probs

    def explore_env(self, observation: List[Dict[str, Union[torch.tensor, int, torch_geometric.data.Data]]]
                    ) -> List[Union[List[torch.tensor], Dict[str, torch.tensor]]] :
        '''
        explore env sampling action from multivariate normal
        '''
        self.actor.eval()
        self.critic.eval()
        actor_action: Dict[str, torch.tensor] = self.actor(observation)  # query the actor network for a mean action.
        # construct sampled action
        sampled_action: Dict[str, torch.tensor] = copy.deepcopy(ACTION_DICT)
        extended_first_node: torch.tensor =  torch.zeros(len(observation["taken_index"])).to(DEVICE)
        extended_first_node[:actor_action["first_node"].shape[0]] = actor_action["first_node"]
        extended_second_node = torch.zeros(len(observation["remain_index"])).to(DEVICE)
        extended_second_node[:actor_action["second_node"].shape[0]] = actor_action["second_node"]
        mean_tensor: torch.tensor = torch.cat([extended_first_node, extended_second_node, actor_action["action_type"]])
        dist_action: MultivariateNormal  = MultivariateNormal(mean_tensor, torch.diag(torch.full(size=(mean_tensor.shape[0],), fill_value=1e-8)).to(DEVICE))
        sampled_cat_action: torch.tensor  = dist_action.sample().to(DEVICE)
        log_prob: torch.tensor =  dist_action.log_prob(sampled_cat_action).to(DEVICE).detach()
        decision: int = sampled_cat_action[len(observation["taken_index"]) + len(observation["remain_index"]):].argmax(dim = 0)
        second_action_dim: torch.tensor = len(observation["remain_index"]) if decision == 0 else len(observation["taken_index"])
        sampled_action["first_node"] = sampled_cat_action[0:len(observation["taken_index"])] if decision != 2 else torch.tensor([]).to(DEVICE)
        sampled_action["second_node"] = sampled_cat_action[len(observation["taken_index"]):len(observation["taken_index"])+second_action_dim] if decision != 2 else torch.tensor([]).to(DEVICE)
        sampled_action["action_type"] = sampled_cat_action[len(observation["taken_index"]) + len(observation["remain_index"]):].to(DEVICE)
        return sampled_action, log_prob

    def compute_rewards_to_go(self, batch_rewards: List[float]) -> torch.tensor:
        '''
        compute rewards to go
        '''
        batch_rewards_to_go: List[float] = []
        # Iterate through each episode backwards to maintain same order
        # in batch_rtgs
        for episode_rewards in reversed(batch_rewards):
            discounted_reward: float = 0 # The discounted reward so far
            for reward in reversed(episode_rewards):
                discounted_reward = reward + discounted_reward * self.config["PPO"]["gamma"]
                batch_rewards_to_go.insert(0, discounted_reward)
        batch_rewards_to_go: torch.tensor = torch.tensor(batch_rewards_to_go, dtype=torch.float).to(DEVICE)    # convert to tensor
        return batch_rewards_to_go

    def evaluate_perf(self, phase: str = "eval") -> List[float]:
        '''
        evaluate performance of actor over eval set
        '''
        self.actor.eval()
        self.env.phase = phase
        performances: List[float] = []    # collect episode performances
        for i in range(len(DwaveEnvironment.eval_data if phase == "train" else DwaveEnvironment.test_data)):
            observation: Dict[str, Union[int, torch.tensor, torch_geometric.data.Data]] = self.env.reset()  # reset env
            for t in range(self.config["PPO"]["max_timesteps_per_episode"]): # collect trajectory
                force_termination: bool = False
                if t == self.config["PPO"]["max_timesteps_per_episode"] - 1:
                    force_termination = True
                actor_action: Dict[str, torch.tensor] = self.actor(observation)  # query the actor network for a mean action.
                observation, reward, done = self.env.step(actor_action, force_termination) # action
                if done is not None:
                  performances.append(done)
                  break
        avg_perf: float = sum(performances)/len(performances)   # compute avg performances
        stdev: float  = np.std(performances)
        return avg_perf, stdev

    def collect_trajectories(self) -> List[Union[List[int], List[float], List[torch.tensor], torch.tensor, List[Dict[str, torch.tensor]], List[Dict[str, Union[int, torch.tensor, torch_geometric.data.Data]]]]]:
        '''
        create a batch of trajectories
        '''
        # Batch data
        batch_observations: List[Dict[str, Union[int, torch.tensor, torch_geometric.data.Data]]] = []   # batch observations
        batch_actions: List[Dict[str, torch.tensor]] = []   # batch actions
        batch_log_probs: List[torch.tensor] = []   # log probs of each action
        batch_rewards: List[float] = []  # batch rewards
        batch_lens: List[int] = []  # episodic lengths in batch
        t: int = 0
        while t < self.config["PPO"]["total_number_timesteps"]:
            episode_rewards: List[float] = []    # collect episode rewards
            observation: Dict[str, Union[torch.tensor, int, torch_geometric.data.Data]] = self.env.reset()  # reset env
            for episode_timestep in range(self.config["PPO"]["max_timesteps_per_episode"]): # collect trajectory
                # Increment timesteps ran this batch so far
                t += 1
                # collect observation
                batch_observations.append(observation)
                action, sampled_action_log_prob = self.explore_env(observation)
                observation, reward, done = self.env.step(action) # action
                # collect reward, action, and log prob
                episode_rewards.append(reward)
                batch_actions.append(action)
                batch_log_probs.append(sampled_action_log_prob)
                if done is not None:
                  break
            # save episode len and reward
            batch_lens.append(episode_timestep + 1) # all time steps in episode (+1 because ep_t start 0)
            batch_rewards.append(episode_rewards) # all rewards in episode
        batch_rewards_to_go: torch.tensor = self.compute_rewards_to_go(batch_rewards)   # compute rewards-to-go (point 4)
        return batch_observations, batch_actions, batch_log_probs, batch_rewards_to_go, batch_lens
```

```python
if 'Callback' in locals():  # check if SMAC libraries are imported (uncommented)
    class CustomCallback(Callback):
        '''
        Custom callback to better print BOHB result

        .. .

        Class Variables:
        ----------
            None

        Instance Variables:
        ----------
            _trial_counter: int
                -- trial so far


        Public Methods:
        ----------
            on_start(self, smbo) -> None
                -- message at start of training

            on_tell_end(self, smbo, info: TrialInfo, value: TrialValue) -> bool | None:
                -- message at end of each trial test


        '''
        def __init__(self) -> None:
            self._trials_counter: int = 0

        def on_start(self, smbo) -> None:
            """
            at start of optimization process
            ...

            Keyword Arguments:
            ----------

              smbo
                 -- optimizer


            Returns
            ----------
              None

            """
            print("Start optimizing")
            print("**************************************************************")
            print("\n")

        def on_tell_end(self, smbo, info: TrialInfo, value: TrialValue) -> bool | None:
            """
            at end of testing configuration
            ...

            Keyword Arguments:
            ----------

              smbo
                 -- optimizer

              info: TrialInfo
                  -- configuration tested

              value: TrialValue
                  -- result of configuration

            Returns
            ----------
              None

            """
            self._trials_counter += 1
            print("****************************************************************")
            print(f"Evaluated {self._trials_counter} trials so far.")
            print("****************************************************************")
            print("\n")
            incumbent = smbo.intensifier.get_incumbent()
            if incumbent is not None:
                print(f"Current incumbent: {dict(incumbent)}")
                print(f"Current incumbent value: {smbo.runhistory.get_cost(incumbent)}")
                print("-------------------------------------------------------------")
                print("\n")
            return None
```

```python
def optimize_net(opt) -> None:
    """
    wrapper for optimization process in BOHB setting plus priors
    ...

    Keyword Arguments:
    ----------

      optimizer
         -- which optimizer employ in finding best parameters

      data: str
         -- if product or real data have to be optimized

    Returns
    ----------
      tensor with shape 1

    """
    default_config = opt.configspace.get_default_configuration()
    print(default_config)
    # Scenario object specifying the optimization environment
    scenario: Scenario = Scenario(
        opt.configspace,
        deterministic = True, # allow deterministic
        n_trials = NUM_TRIALS,  # Evaluate max NUM_TRIALS different trials
        min_budget = 3,  # Train the MLP using a hyperparameter configuration for at least 5 epochs (5 epochs as minimum to foreshadwo tendency)
        max_budget = 10,  # Train the MLP using a hyperparameter configuration for at most 15 epochs (as seen before is higly indicative of performances)
        n_workers = NUM_WORKER_SMAC
        )

    initial_design: MFFacade = MFFacade.get_initial_design(
        scenario,
        additional_configs = [default_config] # We also want to include our default configuration in the initial design
        )

    intensifier: Hyperband = Hyperband(
        scenario,
        incumbent_selection = "highest_budget"
    )
    # define the prior acquisition function, which conduct the optimization using priors over the optimum
    acquisition_function : PriorAcquisitionFunction = PriorAcquisitionFunction(
        acquisition_function = MFFacade.get_acquisition_function(scenario),
        decay_beta = scenario.n_trials / 10,  # Proven solid value by SMAC3
    )
    # Use SMAC to find the best configuration/hyperparameters
    smac: MFFacade = MFFacade(
        scenario,
        opt.compute,
        initial_design = initial_design,
        intensifier = intensifier,
        overwrite = True,  # Overrides any previous results that are found that are inconsistent with the meta-data
        callbacks = [CustomCallback()],  # passing our custom callback
        logging_level = LOGGING_LEVEL_SELECTED,  # all data logging level
        acquisition_function = acquisition_function,
        )

    # optimize
    incumbent = smac.optimize()
    # Get cost of default configuration
    print("****************************************************************")
    print("Test default config")
    print("****************************************************************")
    default_cost = smac.validate(default_config)
    print(f"Default cost: {default_cost}")
    if incumbent is not None:
        # Let's calculate the cost of the incumbent
        print("****************************************************************")
        print("Test final incumbent")
        print("****************************************************************")
        incumbent_cost = smac.validate(incumbent)
        print("\n")
        print(f"Incumbent cost ({intensifier.__class__.__name__}): {incumbent_cost}")
        print("Dictionary parameter")
        print(incumbent.get_dictionary())
        print("\n")
    return None
```

```python
class ParamsOptimizer():
    '''
    handle bohb execution
    '''
    def __init__(
        self,
        task: str = "classification",
        n_fold: int = 5) -> None:
        self.n_fold: int = n_fold
        self.task: str = "classification"
        return None

    #repeatedly called during the optimization and evaluates a given configuration yielding the associated loss.
    def compute(
        self,  config: Configuration,
        seed: int = RANDOM_SEED, budget: int = 10) -> float:
        '''
        compute incumbent performance
        '''
        new_config: Dict[str, Union[float, int, str]] = copy.deepcopy(CONFIG_TEMPLATE_PREDICTOR)
        new_config["batch_size"] = config["batch_size"]
        new_config["gradient_clipping"] = config["gradient_clipping"] if config["gradient_clipping"] != "none" else None
        new_config["scheduler"] = config["scheduler"] if config["scheduler"] != "none" else None
        new_config["scheduler_params"]["T_0"] =  config["T_0"]
        new_config["scheduler_params"]["T_mult"] = config["T_mult"]
        new_config["scheduler_params"]["eta_min"] = config["eta_min"]
        new_config["scheduler_params"]["step_up"] = config["step_up"]
        new_config["scheduler_params"]["step_down"] = config["step_down"]
        new_config["scheduler_params"]["gamma"] = config["gamma"]
        new_config["scheduler_params"]["factor"] = config["factor"]
        new_config["scheduler_params"]["patience"] = config["patience"]
        new_config["loss"]["label_smoothing"] = config["label_smoothing"]
        new_config["loss"]["beta"] = config["beta"]
        new_config["optimizer"]["optimizer_name"] = config["optimizer_name"]
        new_config["optimizer"]["lr"] = config["lr"]
        new_config["optimizer"]["betas"] = (config["beta1"], config["beta2"])
        new_config["optimizer"]["eps"] = config["eps"]
        new_config["optimizer"]["weight_decay"] = config["weight_decay"]
        new_config["optimizer"]["momentum_decay"] = config["momentum_decay"]
        new_config["net"]["hidden_channels"] = config["hidden_channels"]
        new_config["net"]["n_preprocess_layer"] = config["n_preprocess_layer"]
        new_config["net"]["n_message_passing_layer"] = config["n_message_passing_layer"]
        new_config["net"]["n_postprocess_layer"] =  config["n_postprocess_layer"]
        new_config["net"]["normalization"] = config["normalization"]
        new_config["net"]["dropout"] =  config["dropout"]
        new_config["net"]["activation"] =  config["activation"]
        new_config["net"]["prelu_num_parameters"] = 1 if config["prelu_num_parameters"] == "one" else config["hidden_channels"]
        new_config["net"]["prelu_init"] = config["prelu_init"]
        new_config["net"]["aggregation"] = config["aggregation"]
        new_config["net"]["type_conv"] = config["type_conv"]
        new_config["net"]["negative_slope"] = config["negative_slope"]
        new_config["net"]["jumping_knowledge"] = config["jumping_knowledge"]
        # run exp
        test_metric_array: np.ndarray = operative_loop_predictor(
            task = self.task, n_fold = self.n_fold, n_epochs = int(budget) + 1,
            config = new_config, plot = False)
        # SMAC always minimize
        return  (1-test_metric_array.mean())

    #ConfigurationSpace-object defining all hyperparameters, their ranges, and potential dependencies between them.
    @property
    def configspace(self) -> ConfigurationSpace:
         # Build Configuration Space which defines all parameters and their ranges.
        cs: ConfigurationSpace = ConfigurationSpace()
        # place a log-normal prior on the learning rate, so that it is centered on 10^-3,
        #with one unit of standard deviation per multiple of 10 (in log space)
        learning_rate = NormalFloatHyperparameter(
          "lr",
            lower = MIN_LR[self.task],
            upper = MAX_LR[self.task],
            mu = 0.0001, #value used in standard test
            sigma = 0.01,
            log = False,
        )
        cs.add_hyperparameters([learning_rate])
        # label smoothing
        label_smoothing = NormalFloatHyperparameter(
                "label_smoothing",
                lower = 0.0,
                upper = 0.3,
                mu = 0.0, #value used in standard test
                sigma = 0.1,
                log = False
            )
        cs.add_hyperparameters([label_smoothing])
        # beta
        beta = NormalFloatHyperparameter(
                "beta",
                lower = 0.1,
                upper = 2.0,
                mu = 1.0, #value used in standard test
                sigma = 0.1,
                log = False
            )
        cs.add_hyperparameters([beta])
        # gradient clipping
        clip = BetaIntegerHyperparameter(
                    "gradient_clipping", alpha=2, beta=2, lower=1, upper=10)
        cs.add_hyperparameters([clip])
        # Moreover, we believe None is the most likely schedule
        schedule = CategoricalHyperparameter(
            "scheduler",
            ["cosine", "cyclic", "plateau", "none"],
            weights = [1, 1, 1, 4],
            default_value = "none",
        )
        cs.add_hyperparameters([schedule])
        T_0 = UniformIntegerHyperparameter(
            "T_0", lower = 5, upper = 50, default_value = 5)
        T_mult = UniformIntegerHyperparameter(
            "T_mult", lower = 1, upper = 10, default_value = 1)
        eta_min = UniformFloatHyperparameter(
            "eta_min", lower = 0, upper = 0.1, default_value = 0)
        cs.add_hyperparameters([T_0, T_mult, eta_min])
        step_size_up = UniformIntegerHyperparameter(
            "step_up", lower = 1000, upper = 3000, default_value = 2000)
        step_size_down = UniformIntegerHyperparameter(
            "step_down", lower = 1000, upper = 3000, default_value = 2000)
        gamma = UniformFloatHyperparameter(
            "gamma", lower = 0.5, upper = 1.0, default_value = 1.0)
        cs.add_hyperparameters([step_size_up, step_size_down, gamma])
        factor = UniformFloatHyperparameter(
            "factor", lower = 0.01, upper = 0.5, default_value = 0.1)
        patience = UniformIntegerHyperparameter(
            "patience", lower = 1, upper = 20, default_value = 10)
        cs.add_hyperparameters([factor, patience])
        # optimizer
        optimizer_name = CategoricalHyperparameter(
                "optimizer_name",
                ["AdamW", "Adamax", "NAdam", "RAdam"],
                weights = [3, 1, 1, 1],
                default_value = "AdamW",
            )
        cs.add_hyperparameters([optimizer_name])
        # ADAM params
        beta1 = NormalFloatHyperparameter(
            "beta1",
            lower = 0.1,
            upper = 0.999,
            mu =0.9, #value used in standard test
            sigma = 0.1,
            log = False,
            )
        cs.add_hyperparameters([beta1])
        #cond21 = CS.EqualsCondition(beta1, optimizer_name, "Adam")
        #cs.add_condition(cond21)
        beta2 = NormalFloatHyperparameter(
            "beta2",
            lower = 0.1,
            upper = 0.999,
            mu =0.999, #value used in standard test
            sigma = 0.1,
            log = False,
            )
        cs.add_hyperparameters([beta2])
        #cond22 = CS.EqualsCondition(beta2, optimizer_name, "Adam")
        #cs.add_condition(cond22)
        eps = NormalFloatHyperparameter(
            "eps",
            lower = 1e-12,
            upper = 1e-01,
            mu = 1e-08, #value used in standard test
            sigma = 0.1,
            log = False,
            )
        cs.add_hyperparameters([eps])
        #cond23 = CS.EqualsCondition(eps, optimizer_name, "Adam")
        #cs.add_condition(cond23)
        # COMMON params
        #weight decay
        #10-5 a 10-2 o 0
        weight_decay=NormalFloatHyperparameter(
            "weight_decay",
            lower = 0,
            upper = 1e-2,
            mu = 0.0,  # value used in standard test
            sigma = 0.1,
            log = False
        )
        cs.add_hyperparameters([weight_decay])
        # momentum decay
        momentum_decay=NormalFloatHyperparameter(
            "momentum_decay",
            lower = 0.0000004,
            upper = 0.1,
            mu = 0.004,  # value used in standard test
            sigma = 0.01,
            log = False
        )
        cs.add_hyperparameters([momentum_decay])
        # dropout mask
        dropout = NormalFloatHyperparameter(
            "dropout",
            lower = 0.01,
            upper = 0.6,
            mu = 0.1, #value used in standard test
            sigma = 0.2,
            log = False,
            )
        cs.add_hyperparameters([dropout])
        # network specific params
        activation = CategoricalHyperparameter(
                "activation",
                ["relu", "prelu", "swish"],
                weights = [3, 1, 1],
                default_value = "relu",
        )
        cs.add_hyperparameters([activation])
        aggregation = CategoricalHyperparameter(
            "aggregation",
            ["mean", "max", "sum"],
            weights = [3, 1, 1],
            default_value = "mean",
        )
        cs.add_hyperparameters([aggregation])
        normalization = CategoricalHyperparameter(
            "normalization",
            ["batch", "graph", "pair", "mean"],
            weights = [4, 1, 1, 1],
            default_value = "batch",
        )
        cs.add_hyperparameters([normalization])
        type_conv = CategoricalHyperparameter(
            "type_conv",
            ["gat", "reset", "conv", "pdn"],
            weights = [4, 1, 1, 1],
            default_value = "gat",
        )
        cs.add_hyperparameters([type_conv])
        jumping_knowledge = CategoricalHyperparameter(
            "jumping_knowledge",
            ["cat", "max", "lstm"],
            weights = [3, 1, 1],
            default_value = "cat",
        )
        cs.add_hyperparameters([jumping_knowledge])
        n_postprocess_layer = UniformIntegerHyperparameter(
            "n_postprocess_layer", lower = 1, upper = 3, default_value = 3)
        cs.add_hyperparameters([n_postprocess_layer])
        n_message_passing_layer = UniformIntegerHyperparameter(
            "n_message_passing_layer", lower = 1, upper = 3, default_value = 1)
        cs.add_hyperparameters([n_message_passing_layer])
        hidden_channels = UniformIntegerHyperparameter(
            "hidden_channels", lower = 16, upper = 128, default_value = 32)
        cs.add_hyperparameters([hidden_channels])
        prelu_init=NormalFloatHyperparameter(
            "prelu_init",
            lower = 0.0001,
            upper = 0.9999,
            mu = 0.25,  # value used in standard test
            sigma = 0.01,
            log = False
        )
        cs.add_hyperparameters([prelu_init])
        negative_slope=NormalFloatHyperparameter(
            "negative_slope",
            lower = 0.01,
            upper = 0.4,
            mu = 0.2,  # value used in standard test
            sigma = 0.01,
            log = False
        )
        cs.add_hyperparameters([negative_slope])
        prelu_num_parameters = CategoricalHyperparameter(
            "prelu_num_parameters",
            ["one", "all"],
            weights = [2, 1],
            default_value = "one",
        )
        cs.add_hyperparameters([prelu_num_parameters])
        n_preprocess_layer = UniformIntegerHyperparameter(
                  "n_preprocess_layer", lower = 1, upper = 3, default_value = 1)
        cs.add_hyperparameters([n_preprocess_layer])
        # parameter only for predictor setting
        batch_size= BetaIntegerHyperparameter(
                "batch_size", alpha=2, beta=2, lower=16, upper=64)
        cs.add_hyperparameters([batch_size])
        return cs
```

####6.5 DEEPSIG


```python
def compute_aso(ppls_list: List[float],
                confidence_level: float = 0.95,
                seed: int = RANDOM_SEED,
                plot: bool = True) -> float:
    '''
    compute aso confidence score
    ...

    Keyword Arguments:
    ----------
        ppls_list: List[float]
            -- list of performances computed

        confidence_level: float (default: 0.95)
            -- confidence leve

        seed: int (default: RANDOM_SEED)
            -- random seed

        plot: bool (default: True)
            -- if plot result

    Returns
    ----------
        confidence value
    '''
    if plot:
        fig = plt.figure(figsize =(10, 7))
        # Creating axes instance
        ax = fig.add_axes([0, 0, 1, 1])
        # Creating plot
        plt.boxplot((ppls_list))
        # show plot
        plt.show()
    if len(ppls_list)==2:
        min_eps1 = aso(ppls_list[0], ppls_list[1], seed=seed, confidence_level=confidence_level)
        if (min_eps1  < 0.5):
            print("\n")
            print("Using ASO with a confidence level alpha=" + str(1-confidence_level)
                  + "the score distribution of algorithm A based on " + str(len(ppls_list[0]))
                  + "random seeds to be stochastically dominant over B (epsilon_min =" + str(min_eps1)+")")
            return min_eps1
        else:
            min_eps2: float = aso(ppls_list[1], ppls_list[0], seed=seed, confidence_level=confidence_level)
            if (min_eps2  < 0.5):
                print("\n")
                print("Using ASO with a confidence level alpha=" + str(1-confidence_level)
                      + "the score distribution of algorithm B based on " + str(len(ppls_list[0]))
                      + "random seeds to be stochastically dominant over A (epsilon_min =" + str(min_eps2)+")")
            return min_eps2
    else:
        eps_min: float = multi_aso(np.asarray(ppls_list), seed=seed, confidence_level=confidence_level)
        print("\n")
        print("We compared all pairs of models based on " + str(len(ppls_list[0]))
            + " five random seeds each using ASO with a confidence level of alpha=" + str(1-confidence_level)
            + "(before adjusting for all pair-wise comparisons using the Bonferroni correction). Almost stochastic dominance (epsilon_min < tau with tau = 0.5) is indicated in following table.")
        print(eps_min)
        return eps_min
```
